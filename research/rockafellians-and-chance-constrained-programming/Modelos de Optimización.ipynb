# **Problemas de Optimización con Restricciones de Probabilidad (Chance-Constrained Programs)**

En muchos problemas de optimización bajo incertidumbre, las variables aleatorias afectan directamente las restricciones del modelo.  

Una forma de garantizar que las soluciones sean seguras o confiables consiste en exigir que dichas **restricciones se cumplan con alta probabilidad.**  

Esto conduce al planteamiento general del **problema de optimización con restricciones de probabilidad**, también conocido como **Chance-Constrained Program (CCP)** o simplemente aquí **Chance-Constrained**.

---

### **Formulación General**


\begin{aligned}
\min_{x \in \mathbb{R}^n} \quad & c^\top x \\
\text{s.a.} \quad & \mathbb{P}_{\xi}\big[ A(x)\xi \le b(x) \big] \ge 1 - \varepsilon, \\
& x \in X
\end{aligned}


### **Definición de términos**

- $x \in \mathbb{R}^n$: vector de **variables de decisión** (determinísticas).
- $c \in \mathbb{R}^n$: **vector de costos** (coeficientes del objetivo); $c^\top x$ es el valor a minimizar.   
- $\xi \in \mathbb{R}^q$: vector de **parámetros aleatorios** o inciertos.  
- $A(x) \in \mathbb{R}^{m \times q}$, $b(x) \in \mathbb{R}^m$: funciones (en general afines) que definen las restricciones del sistema.  
- $\mathbb{P}_\xi[\cdot]$: medida de **probabilidad** asociada a la distribución de $\xi$.  
- $\varepsilon \in (0,1)$: **nivel de riesgo máximo permitido**; $1 - \varepsilon$ representa el nivel de confianza.  
- $X \subseteq \mathbb{R}^n$: conjunto **determinístico factible** (por ejemplo, cotas, integridad o restricciones adicionales).  

---

### **Interpretación**

- La restricción probabilística  
  $$\mathbb{P}_{\xi}[A(x)\xi \le b(x)] \ge 1 - \varepsilon$$
  asegura que el sistema de desigualdades se cumpla con probabilidad al menos $1 - \varepsilon$.  

- En otras palabras, se permite que las restricciones se violen con probabilidad a lo sumo $\varepsilon$.  

- Este tipo de restricciones permiten **decisiones robustas pero no excesivamente conservadoras**, controlando explícitamente el riesgo de violación.


---

### **Tipos según la estructura del sistema (según Shen y Jiang, 2025)**

En el artículo de Shen y Jiang, los autores distinguen los tipos de restricciones probabilísticas no solo por la cantidad de desigualdades, sino también por **cómo aparece la incertidumbre** en el modelo:

- En el problema **Production Planning (PP)** (que trataremos más adelante), el vector aleatorio $\xi$ está **desacoplado** de las variables de decisión $x$, ya que $A(x)$ es independiente de $x$.  
  Este tipo de restricción se denomina **chance constraint con incertidumbre en el lado derecho (RHS)** (*right-hand side uncertainty*).

- En cambio, en el problema **Hydro Planning (HP)** (que también veremos en detalle), las variables $x$ y $\xi$ están **acopladas** dentro de $A(x)\xi$, lo que genera **incertidumbre en el lado izquierdo (LHS)** (*left-hand side uncertainty*).

Además, según el número de restricciones y la estructura de $A(x)$, se clasifican los *chance constraints* en:

- **Individual**, si $m = 1$.  
- **Two-sided**, si $m = 2$ y las dos filas de $A(x)$ son opuestas (como en el modelo HP).  
- **Joint**, si $m \ge 2$ y las restricciones se deben cumplir simultáneamente (como en el modelo PP).




---


### **Algunos otros tipos de Restricciones de Probabilidad**

- **Individual Chance Constraint:**  
  $\mathbb{P}_{\xi}[a_i(x)^\top \xi \le b_i(x)] \ge 1 - \varepsilon_i, \quad \forall i.$  

- **Joint Chance Constraint:**  
  $\mathbb{P}_{\xi}[A(x)\xi \le b(x)] \ge 1 - \varepsilon.$  

- **Two-Sided Chance Constraint:**  
  $\mathbb{P}_{\xi}[\ell(x) \le g(x,\xi) \le u(x)] \ge 1 - \varepsilon.$  

---

### **Comentarios**

- En general, el conjunto factible de estos problemas **no es convexo**, lo que dificulta su resolución exacta.  (si es importante agregar comentario de Julio)
- Sin embargo, bajo ciertas distribuciones —por ejemplo, *Normal multivariada* o, más generalmente, *log-cóncavas*— la región factible puede ser convexa o aproximarse mediante **restricciones de segundo orden (SOC)**.  
  En particular, Shen y Jiang (2025) demuestran en su Sección 3 que, si la distribución es log-cóncava, el conjunto factible del *Joint Chance Constraint* es convexo.

---
##### *En este trabajo, y en el paper de Shen y Jiang (2025), nos enfocamos exclusivamente en la versión conjunta (Joint Chance Constraint).*

# **Production Planning (PP)** — Chance-Constrained conjunto (Example 1 y Sec. 6.1)

Buscamos adquirir capacidad de producción para que **todas** las demandas se satisfagan con alta probabilidad, minimizando el costo total.

### **Formulación (CC “pura”, usada en nuestra Etapa 1)**
$$
\begin{aligned}
\min_{x \in \mathbb{R}^n} \quad & c^\top x \\
\text{s.a.} \quad & \mathbb{P}\big[\, T x \ge \xi \,\big] \;\ge\; 1-\varepsilon, \\
& 0 \le x_i \le U,\quad i=1,\dots,n.
\end{aligned}
$$

> *Nota:* En el paper, la Sección 6.1 reporta resultados para la versión **pesimista (P-CC)**  
> $$\inf_{Q\in\mathcal P} Q[Tx\ge \xi] \ge 1-\varepsilon,$$
> con el mismo costo y cotas. Aquí partimos con la **CC estándar** (sin ambigüedad), y más adelante podremos comparar.

---

### **Definición de variables y parámetros**

- **Decisión**
  - $x \in \mathbb{R}^n_{\!+}$: capacidades a adquirir en cada instalación (una por instalación).
- **Costo**
  - $c \in \mathbb{R}^n_{\!+}$: costos unitarios de capacidad; el objetivo es $c^\top x$.
- **Cobertura**
  - $T \in \{0,1\}^{m\times n}$: matriz de cobertura/servicio.
    - $T_{ij}=1$ si la instalación $j$ **puede** servir a la ubicación de demanda $i$; $0$ en caso contrario.
- **Demanda aleatoria**
  - $\xi \in \mathbb{R}^m$: vector de demandas por ubicación (variable aleatoria).
  - **Supuesto para simulaciones (Etapa 1):** $\xi \sim \mathcal N(\mu,\Sigma)$ (Normal multivariada), como en el paper.
- **Nivel de servicio**
  - $\varepsilon \in (0,1)$: tolerancia al riesgo; $1-\varepsilon$ es el nivel de confianza conjunto.
- **Cotas**
  - $U>0$: cota superior homogénea por instalación ($0\le x_i\le U$).
- **Dimensiones**
  - $m$: número de ubicaciones de demanda.
  - $n$: número de instalaciones.

---

### **Notas de estructura (coinciden con el paper)**
- **RHS uncertainty:** la incertidumbre está en el **lado derecho** ($Tx \ge \xi$) porque $A(x)$ **no depende de** $x$.  
- **Conjunto (joint) CCP:** la probabilidad es sobre la **satisfacción simultánea** de **todas** las $m$ desigualdades.

## **Estrategia de solución: Método 1: Monte Carlo Sampling**

Queremos encontrar el vector $x$ que **minimiza el costo total** cumpliendo con:
$$
\mathbb{P}\big[\, T x \ge \xi \,\big] \;\ge\; 1 - \varepsilon .
$$
Esto equivale a garantizar que **todas las demandas sean satisfechas simultáneamente con alta probabilidad**.

---

### **Caso A — Distribución Normal multivariada**

Supondremos que $\xi \sim \mathcal{N}(\mu,\Sigma)$.

- Como $T x$ es determinístico, se tiene:
  $$
  T x - \xi \;\sim\; \mathcal{N}(T x - \mu,\;\Sigma),
  $$
  y el evento $\{T x \ge \xi\}$ equivale a $\{T x - \xi \ge 0\}$ (componente a componente).

- **Enfoque base (que usaremos):**  
  **Verificación Monte Carlo** — estimar la probabilidad conjunta por simulación (ver más abajo).

- **(Opcional, como benchmark teórico):**  
  Si quisiéramos una aproximación determinista y conservadora, podríamos usar la **desigualdad de Boole (Bonferroni)** para reemplazar la restricción conjunta por un conjunto de restricciones individuales:
  $$
  \sum_{i=1}^m \varepsilon_i \le \varepsilon, \qquad
  \mathbb{P}[t_i^\top x \ge \xi_i] \ge 1-\varepsilon_i \;\; \forall i.
  $$
  En el caso Normal univariante $\xi_i \sim \mathcal{N}(\mu_i,\sigma_i^2)$, cada restricción individual es equivalente a:
  $$
  t_i^\top x \ge \mu_i + \sigma_i\,\Phi^{-1}(1-\varepsilon_i),
  $$
  donde $t_i^\top$ es la $i$-ésima fila de $T$, $\sigma_i^2=\Sigma_{ii}$ y $\Phi^{-1}$ es el cuantil Normal estándar.  
  > En PP (incertidumbre en el lado derecho) esto lleva a restricciones **lineales** fáciles de resolver, aunque **más conservadoras** que la restricción conjunta real.

- **Evaluar exactamente la probabilidad conjunta** implica la CDF normal multivariada de dimensión $m$; es costoso y no tendremos forma cerrada general. Usaremos Monte Carlo.

---
### **Caso B — Distribución general (no necesariamente Normal)**

Si $\xi$ no es Normal (por ejemplo, uniforme, t-Student, empírica o mezcla de distribuciones), no podemos usar cuantiles analíticos.  
En ese caso, la estrategia sigue siendo la misma: **verificar la probabilidad conjunta por simulación Monte Carlo** o métodos de escenarios.

---

### **Estimación Monte Carlo de la probabilidad conjunta**

- Generamos $N$ muestras independientes $\{\xi^{(1)},\dots,\xi^{(N)}\}$ de la distribución de $\xi$ (Normal u otra).  

- Para un candidato $x$, calculamos la **tasa de incumplimiento conjunta** como:
  $$
  \widehat{p}_{\text{incumplimiento}}(x)
  =
  \frac{1}{N}\sum_{k=1}^N
  \mathbf{1}\!\left[
  \exists\, i \in \{1,\dots,m\}:\; (T x - \xi^{(k)})_i < 0
  \right].
  $$

- Interpretación:  
  - Cada muestra $\xi^{(k)}$ representa un escenario de demanda.  
  - El indicador vale 1 si en ese escenario **alguna demanda no se cumple**.  
  - Por tanto, $\widehat{p}_{\text{incumplimiento}}(x)$ estima la **frecuencia empírica** de incumplimiento del sistema.

- Consideramos $x$ **factible** si
  $$
  \widehat{p}_{\text{incumplimiento}}(x) \le \varepsilon.
  $$

---

### **Resumen operativo**

1. Definir los datos $(T, c, \mu, \Sigma, U, \varepsilon)$ si $\xi$ es Normal; o el generador de muestras si $\xi$ sigue otra distribución.  
2. Implementar la verificación de factibilidad **conjunta** usando Monte Carlo.  
3. Resolver el problema de *chance constraint* estándar (sin ambigüedad Wasserstein) bajo la distribución elegida.

## **Notas sobre el caso conjunto bajo distribución Normal**

Esta sección explica por qué **no podemos calcular directamente** la probabilidad conjunta
$\mathbb{P}[T x \ge \xi]$ usando las tablas típicas de la Normal(0,1),
y por qué recurrimos a la simulación Monte Carlo.

---

### **1️⃣ Caso de una sola restricción**

Supongamos una restricción individual:
$$
\mathbb{P}\big[t_i^\top x \ge \xi_i\big] \ge 1 - \varepsilon_i,
$$
donde $\xi_i \sim \mathcal{N}(\mu_i, \sigma_i^2)$.

Entonces:
$$
t_i^\top x - \xi_i \sim \mathcal{N}(t_i^\top x - \mu_i, \sigma_i^2).
$$

Definimos la variable tipificada:
$$
Z_i = \frac{t_i^\top x - \xi_i}{\sigma_i} \sim
\mathcal{N}\!\left(\frac{t_i^\top x - \mu_i}{\sigma_i}, 1\right).
$$

Y la probabilidad se calcula como:
$$
\mathbb{P}(t_i^\top x \ge \xi_i)
= \mathbb{P}(Z_i \ge 0)
= \Phi\!\left(\frac{t_i^\top x - \mu_i}{\sigma_i}\right),
$$
donde $\Phi(\cdot)$ es la función de distribución de la Normal estándar.

Así obtenemos la forma determinista:
$$
t_i^\top x \ge \mu_i + \sigma_i\,\Phi^{-1}(1-\varepsilon_i),
$$
que es **lineal** y se puede resolver fácilmente.

> ✅ En este caso **sí podemos usar las tablas de la Normal(0,1)** o la función $\Phi^{-1}$.

---

### **2️⃣ Caso de múltiples restricciones conjuntas**

Ahora consideremos la restricción conjunta:
$$
\mathbb{P}[T x \ge \xi] \ge 1 - \varepsilon,
$$
donde $T x \in \mathbb{R}^m$ y $\xi \sim \mathcal{N}(\mu, \Sigma)$.

Esto significa:
$$
\mathbb{P}\big( (T x - \xi)_1 \ge 0, \dots, (T x - \xi)_m \ge 0 \big) \ge 1 - \varepsilon.
$$

El vector $T x - \xi$ tiene una **distribución Normal multivariada**:
$$
T x - \xi \sim \mathcal{N}(T x - \mu, \Sigma),
$$
y sus componentes pueden estar **correlacionadas** (según $\Sigma$).

Entonces la probabilidad que queremos calcular es:
$$
\mathbb{P}[T x \ge \xi]
= \int_{z_1 \ge 0} \cdots \int_{z_m \ge 0}
\phi_m(z; T x - \mu, \Sigma)\, dz_1 \cdots dz_m,
$$
donde $\phi_m(\cdot)$ es la densidad de la Normal multivariada.

> ⚠️ Esta integral **no tiene forma cerrada general** para $m > 1$  
> (solo se puede resolver numéricamente o por simulación).

---

### **3️⃣ Por qué no basta con “llevarla a la Normal estándar”**

Podemos estandarizar el vector:
$$
Z = \Sigma^{-1/2}(\xi - \mu) \sim \mathcal{N}(0, I_m),
$$
de modo que:
$$
T x \ge \xi
\;\Longleftrightarrow\;
\Sigma^{-1/2}(T x - \mu) \ge Z.
$$

Sin embargo, la probabilidad se convierte en:
$$
\mathbb{P}\big(Z_i \le (\Sigma^{-1/2}(T x - \mu))_i,\; i=1,\dots,m\big),
$$
que sigue siendo una **probabilidad conjunta de dimensión $m$**.  
No existen tablas ni fórmulas cerradas para esta CDF multivariada.

---

### **4️⃣ Qué hacemos entonces**

- Si $m$ es pequeño (2 o 3), se puede calcular la CDF multivariada con integración numérica (por ejemplo, con `scipy.stats.multivariate_normal.cdf`), aunque es lento.  
- Si $m$ es mediano o grande (como aquí $m=5$), usamos **simulación Monte Carlo**:
  1. Generamos muchas muestras de $\xi \sim \mathcal{N}(\mu,\Sigma)$.
  2. Contamos cuántas cumplen $T x \ge \xi$.
  3. Estimamos la probabilidad como el porcentaje de cumplimiento.

---

### **5️⃣ Resumen**

| Caso | Distribución | Cálculo | Método |
|------|---------------|----------|---------|
| 1 restricción | Normal univariante | Usa tabla o cuantil estándar $\Phi^{-1}$ | Fórmula exacta |
| m restricciones independientes | Normal multivariante diagonal | Se separan como m casos individuales | Exacto pero conservador |
| m restricciones correlacionadas | Normal multivariante general | Requiere integral multidimensional | Numérico o Monte Carlo |

---

> 💡 En resumen:  
> Las “tablas de la Normal estándar” sólo sirven cuando hay **una variable**.  
> Para el caso conjunto (varias restricciones simultáneas), **no hay una tabla equivalente** y debemos estimar la probabilidad por métodos numéricos o simulación.

## **Implementación del Caso 1 — Generación de datos y simulación Monte Carlo**

Simularemos y resolveremos el problema bajo distribución Normal multivariada, estimando la probabilidad de cumplimiento conjunto:
$$
\mathbb{P}\big[T x \ge \xi\big] \ge 1 - \varepsilon .
$$

---

### **Estructura del código**

**1. Generación de datos base**  
Se fijan $n$, $m$, los costos $c$, la matriz $T$, la media $\mu$ y la covarianza $\Sigma$.  
Se incluye una opción de correlación entre demandas y se fija una semilla (`seed`) para reproducibilidad.  
Cada demanda tiene al menos una instalación que la puede cubrir.

---

**2. Muestreo y verificación de factibilidad**  
Definimos una función para generar muestras $\{\xi^{(1)}, \dots, \xi^{(N)}\}$ de $\mathcal{N}(\mu,\Sigma)$.

Para un candidato $x$, calculamos la **tasa de incumplimiento conjunta**:

$$
\widehat{p}_{\text{incumplimiento}}(x)
=
\frac{1}{N}\sum_{k=1}^N
\mathbf{1}\!\left[
\exists\, i \in \{1,\dots,m\}:\; (T x - \xi^{(k)})_i < 0
\right].
$$

Esta cantidad representa el porcentaje de escenarios en los que **alguna demanda no se satisface**.  
Consideramos $x$ **factible** si:

$$
\widehat{p}_{\text{incumplimiento}}(x) \le \varepsilon.
$$

---

**3. Validación inicial**  
Probamos con $x = (U,\dots,U)$ para verificar que el muestreo y la verificación funcionen correctamente.

---

### **Parámetros configurables**

- $\varepsilon$: nivel de riesgo tolerado (por defecto 0.10 → 90 % de servicio).  
- $N$: número de muestras Monte Carlo (por ejemplo, 5 000 o 20 000).  
- $\rho$: correlación entre demandas (opcional; por defecto sin correlación).  
- Semilla (`seed`): asegura reproducibilidad de resultados.

```python
# --- DATOS Y GENERADORES ---
# --- Setup reproducible ---
import numpy as np

rng = np.random.default_rng(seed=12345)

# --- Parámetros básicos del Caso 1 ---
n = 10            # instalaciones
m = 5             # demandas
U = 200.0         # cota superior por instalación
eps = 0.10        # 1 - eps = 0.90 nivel de servicio

# Costos: c ~ Uniform{1,...,10}
c = rng.integers(low=1, high=11, size=n).astype(float)

# Matriz de cobertura T (0/1) con densidad de unos d
d = 0.5  # densidad objetivo de unos (ajustable)
T = (rng.random((m, n)) < d).astype(float)

# Aseguramos al menos un 1 por fila (factibilidad básica)
for i in range(m):
    if T[i].sum() == 0:
        j = rng.integers(0, n)
        T[i, j] = 1.0

# --- Demanda: Normal multivariada ---
mu = np.array([100, 120, 80, 90, 110], dtype=float)  # ejemplo
sigma = 15.0

# Opción 1: covarianza diagonal (sin correlación)
Sigma = (sigma**2) * np.eye(m)

# Opción 2 (para más adelante): activar correlación rho
use_correlation = False
rho = 0.3  # correlación común
if use_correlation:
    Sigma = (sigma**2) * (rho * np.ones((m, m)) + (1 - rho) * np.eye(m))

# --- Utilidades ---
U_vec = np.full(n, U, dtype=float)

print("Resumen de parámetros:")
print(f"n={n}, m={m}, U={U}, 1-eps={1-eps}")
print("c:", c)
print("T (filas=m, columnas=n):\n", T)
print("mu:", mu)
print("Sigma (diag):", np.diag(Sigma))
```

```python
# --- MONTE CARLO: muestreo y verificación conjunta ---

def sample_xi(N, mu, Sigma, rng=None):
    """
    Genera N muestras de xi ~ N(mu, Sigma).
    Retorna un array (N, m).
    """
    if rng is None:
        rng = np.random.default_rng()
    return rng.multivariate_normal(mean=mu, cov=Sigma, size=N)

def tasa_incumplimiento(x, T, xi_samples):
    """
    x: (n,), T: (m,n), xi_samples: (N,m)
    Devuelve la proporción de escenarios donde AL MENOS una demanda no se satisface,
    es decir, algún i con (T x)_i < (xi)_i.
    """
    # Cobertura determinística por demanda (m,)
    cover = T @ x
    # Compara contra cada muestra (N,m)
    incumple_mat = (cover.reshape(1, -1) < xi_samples)  # True si no se satisface esa demanda
    incumple_esc = incumple_mat.any(axis=1)             # True si falla alguna demanda en el escenario
    return incumple_esc.mean()

def es_factible_mc(x, T, mu, Sigma, eps, N=20000, rng=None):
    """
    Chequea factibilidad probabilística conjunta por Monte Carlo:
    devuelve (es_factible, p_incumpl).
    """
    xi_s = sample_xi(N, mu, Sigma, rng)
    p_hat = tasa_incumplimiento(x, T, xi_s)
    return (p_hat <= eps), p_hat

# Prueba rápida con x = U (todo al máximo)
x_test = U_vec.copy()
N_test = 5000
factible, p_inc = es_factible_mc(x_test, T, mu, Sigma, eps, N=N_test, rng=rng)
print(f"x_test = U → factible={factible}, tasa_incumplimiento≈{p_inc:.4f} (N={N_test})")
```

## **Optimización del problema de Production Planning**

A continuación, implementaremos un procedimiento simple para **encontrar el vector $x^*$ que minimiza el costo total** cumpliendo la restricción probabilística:
$$
\mathbb{P}[T x \ge \xi] \ge 1 - \varepsilon.
$$

El método combina una **resolución determinista base** con una **búsqueda Monte Carlo** sobre un parámetro de seguridad $q$:

1. **LP determinista base:**  
   Se resuelve $\min c^\top x$ s.a. $T x \ge \mu$, $0 \le x \le U$ para obtener un punto inicial $x_0$.

2. **Margen de seguridad ($q$):**  
   Se resuelve el mismo LP con demanda incrementada por $q \cdot s$, donde $s_i = \sqrt{\Sigma_{ii}}$ son las desviaciones típicas.

3. **Búsqueda por bisección:**  
   - Si el $x$ obtenido no cumple la probabilidad requerida, se aumenta $q$.  
   - Si es más conservador de lo necesario, se reduce $q$.  
   - Se repite hasta que $\widehat{p}_{\text{incumplimiento}}(x) \approx \varepsilon$.

4. **Resultado:**  
   - Vector $x^*$ (capacidad óptima).  
   - Costo total $c^\top x^*$.  
   - Probabilidad empírica de servicio $1 - \widehat{p}_{\text{incumplimiento}}(x^*)$.

```python
# --- OPTIMIZACIÓN DEL PROBLEMA DE PRODUCTION PLANNING ---
from scipy.optimize import linprog

def resolver_LP_determinista(c, T, mu, s, q, U_vec):
    """
    Resuelve: min c^T x  s.a.  T x >= mu + q*s,  0 <= x <= U.
    Devuelve x y el costo.
    """
    m, n = T.shape
    A_ub = -T                # T x >= rhs  ->  -T x <= -rhs
    b_ub = -(mu + q * s)
    bounds = [(0, U_vec[i]) for i in range(n)]
    res = linprog(c=c, A_ub=A_ub, b_ub=b_ub, bounds=bounds, method='highs')
    if not res.success:
        print("⚠️ LP no convergió en q =", q)
    return res.x, res.fun

def busqueda_q(c, T, mu, Sigma, eps, U_vec, N=5000, q_min=0.0, q_max=5.0, tol=1e-2, rng=None):
    """
    Busca el menor q tal que la solución del LP sea factible con prob >= 1 - eps.
    Usa bisección sobre q.
    """
    s = np.sqrt(np.diag(Sigma))
    q_lo, q_hi = q_min, q_max
    x_best, cost_best, p_best = None, None, None

    for _ in range(20):  # hasta 20 iteraciones de bisección
        q = 0.5 * (q_lo + q_hi)
        x, cost = resolver_LP_determinista(c, T, mu, s, q, U_vec)
        factible, p_inc = es_factible_mc(x, T, mu, Sigma, eps, N=N, rng=rng)

        print(f"q={q:.3f} → p_inc={p_inc:.4f}, factible={factible}, costo={cost:.2f}")

        if factible:
            # guardamos el mejor hasta ahora y reducimos q
            x_best, cost_best, p_best = x, cost, p_inc
            q_hi = q
        else:
            q_lo = q

        # criterio de convergencia
        if (q_hi - q_lo) < tol:
            break

    return x_best, cost_best, p_best, q_hi

# --- Ejecutamos la búsqueda ---
x_opt, cost_opt, p_inc_opt, q_opt = busqueda_q(
    c, T, mu, Sigma, eps, U_vec, N=5000, q_min=0.0, q_max=5.0, tol=1e-2, rng=rng
)

print("\n--- RESULTADOS ---")
print(f"Margen q*: {q_opt:.3f}")
print(f"Costo óptimo: {cost_opt:.2f}")
print(f"Tasa de incumplimiento empírica: {p_inc_opt:.4f}  → Nivel de servicio ≈ {1 - p_inc_opt:.3f}")
print("x* =", np.round(x_opt, 2))
```

## **Método 2 — Aproximación Bonferroni (ICC)**

Una forma estándar de evitar la simulación Monte Carlo es usar la **aproximación de Bonferroni**, también conocida como **individual chance constraint (ICC)**.

En lugar de exigir que *todas* las restricciones se cumplan simultáneamente con probabilidad al menos $1-\varepsilon$, imponemos restricciones **individuales**, cada una con su propio nivel de confianza $1-\varepsilon_i$, cumpliendo:

$$
\sum_{i=1}^{m} \varepsilon_i \le \varepsilon.
$$

De este modo, si cada restricción $t_i^\top x \ge \xi_i$ se cumple con probabilidad $1-\varepsilon_i$, entonces la probabilidad conjunta cumple:

$$
\mathbb{P}[T x \ge \xi] \ge 1 - \varepsilon.
$$

---

### **Caso Gaussiano univariante**

Si $\xi_i \sim \mathcal{N}(\mu_i, \sigma_i^2)$, se tiene:

$$
\mathbb{P}[t_i^\top x \ge \xi_i] \ge 1-\varepsilon_i
\;\Longleftrightarrow\;
t_i^\top x \ge \mu_i + \sigma_i\,\Phi^{-1}(1-\varepsilon_i),
$$

donde $\Phi^{-1}$ es el cuantil de la Normal estándar.

Cada restricción individual se puede convertir en una **desigualdad determinista lineal**, lo que mantiene la **convexidad** del problema y evita el muestreo.

---

### **Problema resultante (determinista)**

El problema global queda formulado como:

$$
\begin{aligned}
\min_{x \in \mathbb{R}^n} \quad & c^\top x \\
\text{s.a.} \quad & T x \ge \mu + \sigma \cdot \Phi^{-1}(1 - \varepsilon/m), \\
& 0 \le x \le U.
\end{aligned}
$$

Este es un **problema lineal (LP)**: se puede resolver de manera exacta y rápida sin recurrir a simulaciones.

---

### **Por qué el método es conservador**

El método Bonferroni es **conservador** porque **asume independencia total** entre las restricciones.  
Cada restricción individual se asegura con probabilidad $1-\varepsilon_i$, pero el evento conjunto
$$
\{t_1^\top x \ge \xi_1, \dots, t_m^\top x \ge \xi_m\}
$$
no siempre tiene probabilidad exactamente igual al producto de las individuales:  
si las demandas están correlacionadas, o si las restricciones se solapan, la desigualdad de Boole (Bonferroni) solo garantiza una **cota inferior** para la probabilidad conjunta:

$$
\mathbb{P}\Big[\bigcap_i A_i\Big] \ge 1 - \sum_i \mathbb{P}[A_i^c].
$$

Por eso, el conjunto factible resultante es **más pequeño** y la solución suele tener **mayor costo**, garantizando un nivel de servicio incluso superior al requerido.  
En otras palabras, *es seguro, pero paga de más*.

---

### **Por qué validamos con Monte Carlo**

Aunque el modelo ICC es puramente determinista, lo **validamos por Monte Carlo** para:

1. **Medir su nivel de servicio real** bajo la distribución multivariada completa $\xi \sim \mathcal{N}(\mu, \Sigma)$.  
   Así podemos comprobar si el método realmente alcanza (o excede) el nivel de confianza deseado $1-\varepsilon$.

2. **Comparar de manera justa** contra los métodos estocásticos (como el Monte Carlo con bisección o el enfoque por escenarios).  
   De esta forma, usamos la misma métrica empírica $\widehat{p}_{\text{incumplimiento}}(x)$ para todos.

3. **Cuantificar el conservadurismo**: si el método ICC produce tasas de incumplimiento muy inferiores a $\varepsilon$, significa que su solución es más segura, pero probablemente más costosa.

---

### **Conclusión**

El método Bonferroni (ICC) ofrece una solución:

- **Determinista y rápida** (solo requiere resolver un LP).  
- **Convexa y sencilla** de implementar.  
- Pero **más conservadora**, especialmente cuando las demandas están correlacionadas.

Por eso, lo usamos como **benchmark** o referencia inferior, comparando luego su desempeño mediante **validación Monte Carlo** frente a métodos más exactos como el enfoque de **búsqueda en $q$** o el **enfoque por escenarios (SAA)**.

---


```python
# --- MÉTODO ALTERNATIVO 1: BONFERRONI (ICC) ---

from scipy.stats import norm
from scipy.optimize import linprog

def solve_icc_bonferroni(eps_value, c, T, mu, Sigma, U_vec):
    """
    Resuelve la aproximación Bonferroni (ICC):
      min c^T x  s.a.  T x >= mu + sigma*Phi^{-1}(1 - eps/m),  0 <= x <= U.
    """
    m, n = T.shape
    eps_i = eps_value / m
    sigma_vec = np.sqrt(np.diag(Sigma))
    rhs = mu + sigma_vec * norm.ppf(1 - eps_i)

    A_ub = -T
    b_ub = -rhs
    bounds = [(0, U_vec[i]) for i in range(n)]
    res = linprog(c=c, A_ub=A_ub, b_ub=b_ub, bounds=bounds, method='highs')

    if not res.success:
        print("⚠️ LP ICC no convergió.")
    return res.x, res.fun

# --- Ejecutamos el método ICC con los mismos parámetros del Caso 1 ---
x_icc, cost_icc = solve_icc_bonferroni(eps, c, T, mu, Sigma, U_vec)

# --- Validamos su nivel de servicio por Monte Carlo ---
fact_icc, p_check_icc = es_factible_mc(x_icc, T, mu, Sigma, eps, N=50000, rng=None)

print("\n--- RESULTADOS ICC ---")
print(f"Costo óptimo (ICC): {cost_icc:.2f}")
print(f"Tasa empírica de incumplimiento: {p_check_icc:.4f}  → Nivel de servicio ≈ {1 - p_check_icc:.3f}")
print("x_ICC =", np.round(x_icc, 2))
```

## **Método 3 — Enfoque por Escenarios (SAA — Sample Average Approximation)**

Otra estrategia para aproximar la restricción probabilística es el **enfoque por escenarios**,
también llamado **Sample Average Approximation (SAA)**.

### **Idea general**

La restricción
$$
\mathbb{P}[T x \ge \xi] \ge 1 - \varepsilon
$$
se sustituye por un conjunto finito de escenarios aleatorios:
$$
T x \ge \xi^{(k)}, \quad k = 1, \dots, N_s,
$$
donde cada $\xi^{(k)}$ es una muestra independiente de la distribución de la demanda.

Así, el problema estocástico se aproxima mediante el LP determinista:
$$
\begin{aligned}
\min_{x \in \mathbb{R}^n} \quad & c^\top x \\
\text{s.a.} \quad & T x \ge \xi^{(k)}, \quad k = 1, \dots, N_s, \\
& 0 \le x \le U.
\end{aligned}
$$

### **Interpretación**

- Cada escenario representa un posible vector de demandas.
- Exigir que $T x \ge \xi^{(k)}$ para todos los $k$ equivale a cubrir **todas** las realizaciones generadas.
- En la práctica, esto garantiza un nivel de servicio al menos tan alto como $1-\varepsilon$, pero con cierto **exceso de conservadurismo** si $N_s$ es grande.

### **Ventajas y desventajas**

| Ventaja | Desventaja |
|:--|:--|
| Problema determinista (LP simple). | Puede ser muy conservador si $N_s$ es alto. |
| Fácil de implementar. | El número de restricciones crece con $N_s$. |
| Permite calibrar $N_s$ para diferentes niveles de confianza. | No controla directamente la probabilidad $\varepsilon$. |

### **Variantes**

Una versión más refinada del enfoque de escenarios introduce **restricciones con variables binarias** para que solo se violen un número limitado de escenarios (aprox. $\varepsilon N_s$), pero eso convierte el problema en un **MILP**.  
Aquí aplicamos la versión **determinista pura**, como primera comparación.

---


```python
# --- MÉTODO ALTERNATIVO 2: SCENARIO APPROACH (SAA) ---

def solve_saa(c, T, mu, Sigma, U_vec, N_s=1000, rng=None):
    """
    Enfoque SAA: genera N_s escenarios de xi y resuelve:
      min c^T x s.a. T x >= xi^(k)  para todo k,  0 <= x <= U.
    """
    if rng is None:
        rng = np.random.default_rng()
    xi_samples = sample_xi(N_s, mu, Sigma, rng)
    rhs = xi_samples.max(axis=0)  # valor máximo por demanda (enfoque conservador)

    # Construimos y resolvemos LP
    A_ub = -T
    b_ub = -rhs
    bounds = [(0, U_vec[i]) for i in range(T.shape[1])]
    res = linprog(c=c, A_ub=A_ub, b_ub=b_ub, bounds=bounds, method='highs')

    if not res.success:
        print("⚠️ LP SAA no convergió.")
    return res.x, res.fun

# --- Ejecutamos el SAA para los mismos parámetros ---
x_saa, cost_saa = solve_saa(c, T, mu, Sigma, U_vec, N_s=1000, rng=rng)

# --- Validamos nivel de servicio por Monte Carlo ---
fact_saa, p_check_saa = es_factible_mc(x_saa, T, mu, Sigma, eps, N=50000, rng=None)

print("\n--- RESULTADOS SAA ---")
print(f"Costo óptimo (SAA): {cost_saa:.2f}")
print(f"Tasa empírica de incumplimiento: {p_check_saa:.4f}  → Nivel de servicio ≈ {1 - p_check_saa:.3f}")
print("x_SAA =", np.round(x_saa, 2))
```

## **Método 4 — Reformulación Gaussiana tipo SOC (Second-Order Cone)**

Cuando la incertidumbre es **gaussiana** y entra **linealmente** en las restricciones,  
un *chance constraint* puede reformularse (bajo supuestos estándar) como una **restricción cónica de segundo orden (SOC)**.

---

### **Plantilla general (una restricción escalar)**

Sea $\xi \sim \mathcal{N}(\mu,\Sigma)$ y consideremos:

$$
\mathbb{P}\big( a(x)^\top \xi \le b(x) \big) \ge 1-\varepsilon,
$$

con $a(x)$ y $b(x)$ **afines** en $x$.  
Entonces, usando propiedades de la Normal multivariada, esta restricción es **equivalente** a:

$$
a(x)^\top \mu \;+\; \Phi^{-1}(1-\varepsilon)\; \big\|\, \Sigma^{1/2} a(x)\,\big\|_2 \;\le\; b(x).
$$

Esta es una **SOC**, porque el término de norma entra como $\| \cdot \|_2$.

> **Intuición:**  
> $a(x)^\top \xi$ es Normal con media $a(x)^\top \mu$ y desviación $\|\Sigma^{1/2} a(x)\|_2$.  
> Exigir probabilidad $\ge 1-\varepsilon$ equivale a exigir que $b(x)$ esté al menos  
> $\Phi^{-1}(1-\varepsilon)$ desviaciones estándar por encima de la media.

---

### **Aplicación a Production Planning (PP) con incertidumbre en el RHS**

En PP exigimos **simultáneamente**:

$$
\mathbb{P}\big( (T x)_i \ge \xi_i,\; \forall i \big) \ge 1-\varepsilon.
$$

Si usamos la **aproximación individual** (asignamos $\varepsilon_i$ y exigimos cada restricción por separado),  
cada restricción toma la forma:

$$
\mathbb{P}\big( \xi_i \le (T x)_i \big) \ge 1-\varepsilon_i
\;\Longleftrightarrow\;
(T x)_i \ge \mu_i + \sigma_i\,\Phi^{-1}(1-\varepsilon_i),
$$

donde $\sigma_i^2 = \Sigma_{ii}$.

Aquí $a(x)$ **no depende de $x$**, porque afecta a $\xi$ como $e_i^\top \xi$.  
Por tanto, la norma $\|\Sigma^{1/2} a(x)\|_2$ se reduce a $\sigma_i$ (un escalar),  
y la SOC **se vuelve lineal**. En suma, para PP (RHS) la reformulación SOC **degenera en un LP**.

> **Clave:**  
> En problemas con incertidumbre en el **LHS** (por ejemplo, $a(x)^\top \xi$ con $a(x)$ dependiente de $x$),  
> el término de norma sí depende de $x$ y la restricción queda como una verdadera **SOC**.  
> En PP (RHS), la misma fórmula produce líneas (es decir, un LP).

---

### **Resumen**

- El enfoque SOC es **exacto** para *chance constraints gaussianas lineales* (a nivel individual).  
- En PP (RHS) se reduce a un **LP** con umbrales basados en cuantiles.  
- Aun así, el marco SOC es útil (y reusable) si luego atacamos modelos con **acoplamiento $x$–$\xi$ (LHS)**,  
  donde sí aparece la **norma** y la **SOC** genuina.

---


```python
# ---- MÉTODO 4 (PP-RHS): Reformulación Gaussiana tipo SOC, con fallback robusto ----
# En PP-RHS, la "SOC" se reduce a restricciones lineales: T x >= mu + sigma * z.
# Esta celda prueba con cvxpy usando el primer solver disponible (OSQP, SCS, ECOS)
# y, si ninguno está, hace fallback a scipy.linprog.

import numpy as np
import cvxpy as cp
from scipy.stats import norm
from scipy.optimize import linprog

def solve_gaussian_soc_pp_robust(c, T, mu, Sigma, U_vec, eps, eps_allocation="uniform"):
    m, n = T.shape

    # eps_i
    if eps_allocation == "uniform":
        eps_i = np.full(m, eps / m)
    else:
        eps_i = np.asarray(eps_allocation, dtype=float)
        assert len(eps_i) == m and eps_i.sum() <= eps + 1e-10, "eps_i inválido"

    sigma = np.sqrt(np.diag(Sigma))
    z = norm.ppf(1 - eps_i)
    rhs = mu + sigma * z  # umbral determinista

    # Intentamos con cvxpy (OSQP -> SCS -> ECOS)
    installed = set(cp.installed_solvers())
    preferred = [s for s in ["OSQP", "SCS", "ECOS"] if s in installed]

    if len(preferred) > 0:
        x = cp.Variable(n, nonneg=True)
        constr = [T @ x >= rhs, x <= U_vec]
        prob = cp.Problem(cp.Minimize(c @ x), constr)
        try:
            prob.solve(solver=preferred[0], verbose=False)
            if x.value is None:
                raise RuntimeError("cvxpy no retornó solución.")
            x_sol, cost_sol = x.value, prob.value
            solved_with = f"cvxpy/{preferred[0]}"
            return x_sol, cost_sol, solved_with
        except Exception as e:
            print(f"⚠️ cvxpy con {preferred[0]} falló: {e}")
            # fallthrough a SciPy

    # Fallback a SciPy (LP lineal equivalente)
    A_ub = -T
    b_ub = -rhs
    bounds = [(0, float(U_vec[i])) for i in range(n)]
    res = linprog(c=c, A_ub=A_ub, b_ub=b_ub, bounds=bounds, method='highs')
    if not res.success:
        raise RuntimeError("Fallback SciPy linprog no convergió. Revisa datos.")
    return res.x, res.fun, "scipy.linprog"

# --- Ejecutar con los datos ya cargados del Caso 1 ---
x_soc, cost_soc, backend = solve_gaussian_soc_pp_robust(c, T, mu, Sigma, U_vec, eps, eps_allocation="uniform")

# --- Validación Monte Carlo (misma métrica que otros métodos) ---
fact_soc, p_check_soc = es_factible_mc(x_soc, T, mu, Sigma, eps, N=50000, rng=None)

print("\n--- RESULTADOS SOC (PP-RHS) ---")
print(f"Backend usado: {backend}")
print(f"Costo óptimo (SOC/LP): {cost_soc:.2f}")
print(f"Tasa empírica de incumplimiento: {p_check_soc:.4f}  → Nivel de servicio ≈ {1 - p_check_soc:.3f}")
print("x_SOC =", np.round(x_soc, 2))
```

## **Caso 2 — Escalabilidad y comparación de métodos**

En este segundo caso aumentamos la dimensión del problema y probamos los **cuatro métodos** desarrollados:

### **Configuración del Caso 2**

- Número de instalaciones: $n = 30$  
- Número de demandas: $m = 20$  
- Capacidades máximas: $U = 200$  
- Costos: $c \sim \text{Uniform}\{1,\dots,10\}$  
- Niveles de riesgo: $\varepsilon \in \{0.10,\;0.15,\;0.20\}$  
- Matriz de cobertura: $T \in \{0,1\}^{m\times n}$, densidad ≈ 0.4  
- Media de demanda: $\mu_i \sim \text{Uniform}[80,120]$  
- Desviación típica: $\sigma_i = 15$  
- Covarianza: $\Sigma = \sigma^2 I_m$ (sin correlación)

Con esta configuración comparamos los cuatro métodos en términos de:

- **Costo óptimo obtenido**,  
- **Tasa empírica de incumplimiento** (estimada por Monte Carlo, N = 20 000),  
- **Nivel de servicio alcanzado** $(1-\widehat p_{\text{inc}})$.

---


```python
# --- CASO 2: NUEVOS PARÁMETROS Y COMPARACIÓN DE MÉTODOS ---

import numpy as np
import pandas as pd

rng = np.random.default_rng(seed=2025)

# --- Parámetros base ---
n, m = 30, 20
U = 200.0
U_vec = np.full(n, U)
eps_list = [0.10, 0.15, 0.20]

# Costos y matriz de cobertura
c = rng.integers(low=1, high=11, size=n).astype(float)
d = 0.4  # densidad de unos
T = (rng.random((m, n)) < d).astype(float)
for i in range(m):
    if T[i].sum() == 0:
        T[i, rng.integers(0, n)] = 1.0

# Media y covarianza de la demanda
mu = rng.uniform(80, 120, size=m)
sigma = 15.0
Sigma = (sigma**2) * np.eye(m)

print(f"Resumen Caso 2: n={n}, m={m}, niveles eps={eps_list}")
print("Promedio de mu:", np.mean(mu).round(2))

# --- Bucle principal ---
results = []
for eps in eps_list:
    print(f"\n=== ε = {eps:.2f} (nivel de servicio {1-eps:.2%}) ===")

    # 1) Monte Carlo + búsqueda en q
    r_mc = {}
    x_mc, cost_mc, p_inc_mc, q_mc = busqueda_q(
        c, T, mu, Sigma, eps, U_vec, N=8000, q_min=0.0, q_max=5.0, tol=1e-2, rng=rng
    )
    fact_mc, p_check_mc = es_factible_mc(x_mc, T, mu, Sigma, eps, N=20000, rng=None)
    r_mc = dict(method="Monte Carlo + bisección", eps=eps, cost=cost_mc, p_inc=p_check_mc, service=1-p_check_mc)

    # 2) Bonferroni (ICC)
    x_icc, cost_icc = solve_icc_bonferroni(eps, c, T, mu, Sigma, U_vec)
    fact_icc, p_check_icc = es_factible_mc(x_icc, T, mu, Sigma, eps, N=20000, rng=None)
    r_icc = dict(method="Bonferroni (ICC)", eps=eps, cost=cost_icc, p_inc=p_check_icc, service=1-p_check_icc)

    # 3) Scenario Approach (SAA)
    x_saa, cost_saa = solve_saa(c, T, mu, Sigma, U_vec, N_s=1500, rng=rng)
    fact_saa, p_check_saa = es_factible_mc(x_saa, T, mu, Sigma, eps, N=20000, rng=None)
    r_saa = dict(method="Scenario Approach (SAA)", eps=eps, cost=cost_saa, p_inc=p_check_saa, service=1-p_check_saa)

    # 4) Reformulación Gaussiana tipo SOC
    x_soc, cost_soc, backend = solve_gaussian_soc_pp_robust(c, T, mu, Sigma, U_vec, eps)
    fact_soc, p_check_soc = es_factible_mc(x_soc, T, mu, Sigma, eps, N=20000, rng=None)
    r_soc = dict(method=f"Gaussian SOC ({backend})", eps=eps, cost=cost_soc, p_inc=p_check_soc, service=1-p_check_soc)

    results.extend([r_mc, r_icc, r_saa, r_soc])

# --- Mostrar resultados comparativos ---
df = pd.DataFrame(results)
df = df.round({"cost": 2, "p_inc": 4, "service": 4})
print("\n--- RESULTADOS COMPARATIVOS CASO 2 ---")
print(df)

# --- Resumen promedio por método ---
summary = df.groupby("method")[["cost","p_inc","service"]].mean().round(4)
print("\n--- RESUMEN PROMEDIO POR MÉTODO ---")
print(summary)
```

- Covarianza: $\Sigma = \sigma^2 I_m$ (sin correlación, para aislar el efecto del tamaño)

**Métrica común de evaluación** (para comparar métodos):  
- **Costo** $c^\top x$  
- **Tasa empírica de incumplimiento** $\widehat p_{\text{inc}}$ (Monte Carlo)  
- **Nivel de servicio** $1-\widehat p_{\text{inc}}$

Métodos a comparar:
1. **Monte Carlo + bisección en $q$** (ajusta para lograr $p_{\text{inc}}\approx \varepsilon$).  
2. **Bonferroni (ICC)** (determinista; conservador).

```python
# === CASO 3: Instancia grande y comparación integral de los 4 métodos ===
# Supone que ya están definidas en el notebook las funciones:
# - busqueda_q, es_factible_mc, solve_icc_bonferroni, solve_saa, solve_gaussian_soc_pp_robust
# - sample_xi (usada por algunos helpers)

import numpy as np
import pandas as pd

rng = np.random.default_rng(seed=303)

# --- Parámetros del Caso 3 ---
n, m = 60, 40
U = 200.0
U_vec = np.full(n, U)
eps_list = [0.05, 0.10, 0.15]

# Costos y matriz de cobertura
c = rng.integers(low=1, high=11, size=n).astype(float)

densidad = 0.35
T = (rng.random((m, n)) < densidad).astype(float)
# Garantizamos al menos una instalación por demanda
for i in range(m):
    if T[i].sum() == 0:
        T[i, rng.integers(0, n)] = 1.0

# Media y covarianza de la demanda
mu = rng.uniform(85, 125, size=m)
sigma = 18.0
Sigma = (sigma**2) * np.eye(m)

print(f"Resumen Caso 3: n={n}, m={m}, eps_list={eps_list}")
print(f"Promedio(mu)={np.mean(mu):.2f},  sigma={sigma},  densidad(T)≈{T.mean():.2f}")

# --- Tamaños de muestreo (equilibrio entre precisión y tiempo) ---
N_mc_inner = 8000    # Monte Carlo dentro de la búsqueda en q
N_mc_check  = 20000  # Revalidación final
N_saa = 2000         # Escenarios para SAA (crea 2000 restricciones por fila)

# --- Ejecución de los 4 métodos para cada epsilon ---
rows = []
for eps in eps_list:
    print(f"\n=== ε = {eps:.2f} (nivel de servicio {1-eps:.2%}) ===")

    # 1) Monte Carlo + bisección en q
    x_mc, cost_mc, p_inc_mc, q_mc = busqueda_q(
        c, T, mu, Sigma, eps, U_vec,
        N=N_mc_inner, q_min=0.0, q_max=6.0, tol=2e-2, rng=rng
    )
    fact_mc, p_check_mc = es_factible_mc(x_mc, T, mu, Sigma, eps, N=N_mc_check, rng=None)
    print(f"[MC]  cost={cost_mc:.2f} | q*={q_mc:.3f} | p_inc≈{p_check_mc:.4f} | factible={fact_mc}")
    rows.append(dict(method="Monte Carlo + bisección", eps=eps,
                     cost=cost_mc, p_inc=p_check_mc, service=1-p_check_mc))

    # 2) Bonferroni (ICC)
    x_icc, cost_icc = solve_icc_bonferroni(eps, c, T, mu, Sigma, U_vec)
    fact_icc, p_check_icc = es_factible_mc(x_icc, T, mu, Sigma, eps, N=N_mc_check, rng=None)
    print(f"[ICC] cost={cost_icc:.2f} | p_inc≈{p_check_icc:.4f} | factible={fact_icc}")
    rows.append(dict(method="Bonferroni (ICC)", eps=eps,
                     cost=cost_icc, p_inc=p_check_icc, service=1-p_check_icc))

    # 3) Scenario Approach (SAA)
    x_saa, cost_saa = solve_saa(c, T, mu, Sigma, U_vec, N_s=N_saa, rng=rng)
    fact_saa, p_check_saa = es_factible_mc(x_saa, T, mu, Sigma, eps, N=N_mc_check, rng=None)
    print(f"[SAA] cost={cost_saa:.2f} | p_inc≈{p_check_saa:.4f} | factible={fact_saa}")
    rows.append(dict(method="Scenario Approach (SAA)", eps=eps,
                     cost=cost_saa, p_inc=p_check_saa, service=1-p_check_saa))

    # 4) Reformulación Gaussiana SOC (LP en PP-RHS) con fallback robusto
    x_soc, cost_soc, backend = solve_gaussian_soc_pp_robust(c, T, mu, Sigma, U_vec, eps)
    fact_soc, p_check_soc = es_factible_mc(x_soc, T, mu, Sigma, eps, N=N_mc_check, rng=None)
    print(f"[SOC] cost={cost_soc:.2f} | backend={backend} | p_inc≈{p_check_soc:.4f} | factible={fact_soc}")
    rows.append(dict(method=f"Gaussian SOC ({backend})", eps=eps,
                     cost=cost_soc, p_inc=p_check_soc, service=1-p_check_soc))

# --- Tabla comparativa ---
df3 = pd.DataFrame(rows)
df3 = df3.round({"cost": 2, "p_inc": 4, "service": 4})
print("\n--- RESULTADOS COMPARATIVOS CASO 3 ---")
print(df3)

# --- Resumen por método ---
summary3 = df3.groupby("method")[["cost","p_inc","service"]].mean().round(4)
print("\n--- RESUMEN PROMEDIO POR MÉTODO (Caso 3) ---")
print(summary3)
```

## **Otras opciones de distribución 1**

Hasta ahora hemos trabajado con el caso **Normal multivariante**, que tiene propiedades analíticas ideales (convexidad, cierre bajo proyecciones lineales, forma explícita de los cuantiles).  
Sin embargo, en muchos contextos reales las variables aleatorias no son normales: pueden ser discretas, asimétricas o tener colas más pesadas.  

A continuación se enumeran diversas familias de distribuciones que podrían usarse para **simular o analizar restricciones de probabilidad**, indicando en cada caso:

- si la convexidad se conserva o no,  
- cómo se suelen manejar numéricamente,  
- y en qué contextos son útiles o realistas.

---

### **1️⃣ Elípticas (generalización de la Normal)**

Incluyen aquellas cuya densidad depende solo de una forma cuadrática:

$$
f(\xi) = g\big((\xi - \mu)^\top \Sigma^{-1} (\xi - \mu)\big)
$$

#### Ejemplos
- **Normal multivariante** (ya usada).  
- **t de Student multivariante** (colas más pesadas, parámetro de grados de libertad $\nu$).  
- **Laplace multivariante** (colas intermedias, más robusta a outliers).  
- **Cauchy multivariante** (colas muy pesadas, sin momentos finitos).

#### Propiedades
- Proyección lineal $a^\top \xi$ mantiene la forma univariada de la misma familia.  
- Se pueden seguir usando restricciones del tipo:
  $$
  a^\top \mu + k_\varepsilon \|\Sigma^{1/2}a\|_2 \le b
  $$
  donde $k_\varepsilon$ es el **cuantil** de la distribución univariada correspondiente.  
- Convexidad: ✅ (mientras la función generadora $g$ sea log-cóncava).  
- Resolución: SOC o LP con diferente cuantil.

#### Casos de uso
- Modelar incertidumbre con colas más gruesas (riesgo financiero, energía, logística).

---

### **2️⃣ Log-cóncavas**

Una densidad $f(\xi)$ es **log-cóncava** si $\log f(\xi)$ es cóncava.  
Este conjunto es muy amplio y garantiza que los conjuntos de probabilidad
$$
\{x:\mathbb{P}[A(x)\xi\le b(x)]\ge 1-\varepsilon\}
$$
sean **convexos** (Shen & Jiang, 2021).

#### Ejemplos
- Normal (ya cubierta)  
- Laplace  
- Exponencial  
- Uniforme sobre un politopo  
- Distribuciones truncadas de las anteriores

#### Propiedades
- Convexidad garantizada (teóricamente).  
- No siempre existe reformulación analítica → se usan Monte Carlo o convexificaciones locales.  
- Se pueden emplear métodos de **gradiente estocástico**, **submuestreo adaptativo**, o **bounds tipo Cantelli/Chebyshev**.

#### Casos de uso
- Modelos donde la incertidumbre es positiva o acotada (demanda, tiempos de proceso, etc.).

---

### **3️⃣ Uniformes (en regiones compactas)**

Si $\xi$ es uniforme en una región poligonal o caja hipercúbica $[a,b]^m$:

$$
\mathbb{P}[A(x)\xi \le b(x)] = 1
$$
si se cumple en todos los vértices extremos.

#### Propiedades
- Equivalente a exigir restricciones deterministas en los vértices.  
- Convexidad: ✅ (por naturaleza lineal).  
- Extremadamente conservador: cubre el peor caso.  
- Ideal para comparación con métodos robustos (min-max).

---

### **4️⃣ Distribuciones exponenciales y Gamma**

Modelan incertidumbre **positiva** y **asimétrica**.

#### Ejemplos
- **Exponencial:** $f(\xi_i)=\lambda e^{-\lambda\xi_i}$  
- **Gamma:** generaliza la anterior con parámetro de forma $k$ y escala $\theta$.

#### Propiedades
- Log-cóncavas (convexas en sentido probabilístico).  
- Reformulación exacta no analítica; se pueden usar aproximaciones basadas en:
  - cuantiles empíricos (Monte Carlo),
  - desigualdades de Cantelli,
  - transformaciones logarítmicas (si se desea simetrizar).  
- Convexidad: ✅ aproximada.

#### Casos de uso
- Modelar demandas, tiempos de servicio o caudales (hidrología).

---

### **5️⃣ Empíricas / de escenarios (no paramétricas)**

Construidas a partir de $N$ muestras $\{\xi^{(k)}\}$ observadas o simuladas.  

$$
\mathbb{P}(\xi=\xi^{(k)}) = \frac{1}{N}
$$

#### Propiedades
- La restricción $\mathbb{P}[A(x)\xi\le b(x)]\ge 1-\varepsilon$  
  se convierte en un conjunto finito de restricciones:
  $A(x)\xi^{(k)}\le b(x)$ en al menos $(1-\varepsilon)N$ escenarios.
- Convexidad: ✅ (en el caso determinista completo).  
- Puede representarse como LP o MILP dependiendo si se permite violar escenarios.  
- Requiere muestreo representativo.

#### Casos de uso
- Datos históricos o simulaciones estocásticas complejas sin distribución conocida.

---

### **6️⃣ Mixturas de distribuciones (multi-modales)**

Combinaciones de $K$ componentes:
$$
f(\xi) = \sum_{k=1}^K \pi_k f_k(\xi)
$$

#### Ejemplos
- Mixturas gaussianas.  
- Mixturas exponenciales.

#### Propiedades
- No log-cóncavas (no convexas).  
- No hay reformulación cerrada → se usa **Monte Carlo**, o bien se modela cada componente por separado y se combinan resultados ponderados:
  $$
  \mathbb{P}[A(x)\xi\le b(x)] = \sum_k \pi_k \mathbb{P}_k[A(x)\xi\le b(x)]
  $$
- Convexidad: ❌  
- Resolución: numérica, basada en simulación.

#### Casos de uso
- Escenarios multimodales (por ejemplo, distintos regímenes de operación o estacionalidad).

---

### **7️⃣ Distribuciones de colas pesadas (riesgo extremo)**

Modelan fenómenos raros pero con alto impacto.

#### Ejemplos
- **Pareto**, **Weibull** (cola derecha larga).  
- **Log-normal**.  
- **Stable (α-estable)**.  
- **Generalized Extreme Value (GEV)**, **Gumbel**, **Fréchet**.

#### Propiedades
- No convexas en general.  
- Sin momentos finitos en algunos casos (ej. Cauchy).  
- Solo abordables mediante **simulación Monte Carlo** o **enfoques robustos** (Wasserstein / Rockafellar-Uryasev).  
- Sensibles a cuantiles altos (pérdidas, catástrofes, energía).

#### Casos de uso
- Riesgo financiero, seguros, modelado de picos de demanda o fallas catastróficas.

---

### **8️⃣ Distribuciones discretas / categóricas**

Cuando las incertidumbres toman valores finitos o discretos:
- Bernoulli, Binomial, Poisson.  
- Distribuciones empíricas discretas con pesos.

#### Propiedades
- Convexidad: depende de la estructura.  
- Resolución típica: enumerar o muestrear escenarios posibles.  
- Si el número de combinaciones es grande, se puede usar:
  - *Scenario reduction*,  
  - *sampling adaptativo*,  
  - *robust bounds* sobre las probabilidades.  

#### Casos de uso
- Fallas de componentes, demanda en unidades enteras, decisiones lógicas.

---

### **9️⃣ Truncadas o acotadas**

Versiones de distribuciones continuas restringidas a un intervalo $[a,b]$.

#### Ejemplos
- Normal truncada.  
- t truncada.  
- Uniforme restringida.

#### Propiedades
- Siguen siendo log-cóncavas.  
- Requieren integrar sobre una región acotada.  
- Reformulación cerrada rara → se maneja por simulación.  
- Convexidad: ✅ en sentido general.

#### Casos de uso
- Incertidumbres físicamente acotadas (capacidad, caudales, potencia).

---

### **🔟 Distribuciones correlacionadas o dependientes**

Hasta ahora asumimos independencia entre componentes de $\xi$.  
En muchos casos hay correlación, que puede modelarse con:

- **Copulas** (p.ej. Gaussiana, t-Copula, Clayton, Gumbel).  
- **Modelos vine copula** (dependencia jerárquica).  

#### Propiedades
- Permiten combinar cualquier marginal $f_i(\xi_i)$ con una estructura de dependencia.  
- Convexidad: ❌ en general.  
- Resolución: simulación Monte Carlo con muestreo correlacionado.  

#### Casos de uso
- Modelos de energía, clima, mercados financieros.

---

### **Resumen general**

| Familia | Ejemplos | Convexidad | Resolución típica | Comentario |
|----------|-----------|-------------|-------------------|-------------|
| Elípticas | Normal, t, Laplace | ✅ | SOC/LP | Reformulación directa posible |
| Log-cóncavas | Exp, Uniforme, Truncadas | ✅ | Convexificación o MC | Amplia y práctica |
| Uniformes | Cajas, politopos | ✅ | Peor caso determinista | Conservador |
| Empíricas | Escenarios discretos | ✅ | LP/MILP | Basado en datos |
| Mixturas | Gaussian mixtures | ❌ | MC | Multimodal |
| Colas pesadas | Pareto, Weibull, GEV | ❌ | MC/Robusto | Riesgo extremo |
| Discretas | Bernoulli, Poisson | ⚠️ | Enumeración o MILP | Frecuente en logística |
| Truncadas | Normal truncada | ✅ | MC | Realistas en límites físicos |
| Dependientes | Copulas | ❌ | MC | Control de correlaciones |

---

**Conclusión:**
- Las familias **elípticas** y **log-cóncavas** permiten reformulaciones o aproximaciones convexas y son ideales para simulaciones controladas.  
- Las distribuciones **no convexas o multimodales** requieren simulación pura o enfoques robustos.  
- A nivel práctico, vale la pena probar al menos:
  - Normal multivariante (base),  
  - t multivariante (colas pesadas),  
  - Uniforme en caja (peor caso),  
  - Exponencial o Gamma (asimetría positiva),  
  - Empírica (datos/simulación).  

Estas cubren la mayoría de los comportamientos de incertidumbre observables en sistemas reales.

## **Otras opciones de distribución 2**

Hasta ahora hemos trabajado con la **Normal multivariante**, ideal para formular problemas convexos.  
Pero existen muchas otras distribuciones clásicas de teoría de probabilidades que pueden ser útiles para representar diferentes tipos de incertidumbre en problemas reales.

A continuación se presenta una guía de familias **continuas y discretas**, indicando:

- qué representan,  
- qué propiedades tienen (colas, soporte, simetría),  
- si conservan convexidad o se resuelven numéricamente,  
- y ejemplos de aplicación o interpretación en optimización estocástica.

---

## 🔹 **Distribuciones continuas más comunes**

### **1️⃣ Uniforme (rectangular)**

#### Definición:
$$
\xi \sim \text{Uniform}(a,b)
$$

#### Propiedades:
- Soporte finito $[a,b]$  
- Densidad constante.  
- Varianza moderada, sin colas.  
- Log-cóncava → convexidad garantizada.

#### En chance constraints:
- La probabilidad de violación se puede calcular directamente.  
- En vectorial, basta con chequear las restricciones en los **vértices extremos** del hiper-rectángulo.

#### Uso típico:
- Incertidumbre acotada o peor caso dentro de rangos.  
- Ejemplo: demanda que siempre está entre 80 y 120.

---

### **2️⃣ Exponencial**

#### Definición:
$$
f(x;\lambda) = \lambda e^{-\lambda x}, \quad x \ge 0
$$

#### Propiedades:
- Cola derecha larga, solo valores positivos.  
- Log-cóncava → convexidad mantenida.  
- Cuantil analítico: $x_p = -\frac{1}{\lambda}\ln(1-p)$.

#### En optimización:
- Modela tiempos de espera, caudales o demandas positivas.  
- Se puede linealizar mediante aproximación Monte Carlo o mediante desigualdades tipo Cantelli.

---

### **3️⃣ Gamma**

#### Definición:
$$
f(x;k,\theta) = \frac{1}{\Gamma(k)\theta^k} x^{k-1} e^{-x/\theta}, \quad x \ge 0
$$

#### Propiedades:
- Generaliza la Exponencial (cuando $k=1$).  
- Controla la forma y dispersión.  
- Log-cóncava si $k \ge 1$.  
- Cuantil se obtiene numéricamente o vía `scipy.stats.gamma.ppf`.

#### Aplicación:
- Modelar demanda agregada, flujos, tiempos de producción.  
- Puede representarse como suma de exponenciales independientes.

---

### **4️⃣ Beta**

#### Definición:
$$
f(x;\alpha,\beta) = \frac{x^{\alpha-1}(1-x)^{\beta-1}}{B(\alpha,\beta)}, \quad 0<x<1
$$

#### Propiedades:
- Soporte limitado (0,1).  
- Flexible: puede ser simétrica o muy asimétrica.  
- Log-cóncava si $\alpha,\beta \ge 1$.  
- Convexa en sentido probabilístico.  

#### Uso:
- Modelar proporciones o rendimientos (p.ej., eficiencia de una planta, porcentaje de energía almacenada).  
- Restricciones se pueden manejar con muestreo o transformación afín $y=a+(b-a)x$.

---

### **5️⃣ Weibull**

#### Definición:
$$
f(x;k,\lambda) = \frac{k}{\lambda} \left(\frac{x}{\lambda}\right)^{k-1} e^{-(x/\lambda)^k}, \quad x>0
$$

#### Propiedades:
- Colas controladas por $k$:
  - $k<1$: cola pesada,
  - $k>1$: más ligera.  
- Log-cóncava para $k\ge 1$.  
- Cuantiles conocidos (función inversa de $F$ analítica).

#### Aplicación:
- Modelar vida útil, fallas, duraciones.  
- Convexa si $k\ge1$.

---

### **6️⃣ Log-normal**

#### Definición:
Si $Y \sim \mathcal{N}(\mu,\sigma^2)$, entonces $X = e^Y \sim \text{LogNormal}(\mu,\sigma)$.

#### Propiedades:
- Positiva, fuertemente asimétrica, cola pesada.  
- No log-cóncava.  
- Convexidad ❌, requiere simulación Monte Carlo.

#### Aplicación:
- Modelar precios, ingresos, caudales, tiempos.  
- Muy común en finanzas y energía.

---

### **7️⃣ Pareto**

#### Definición:
$$
f(x;x_m,\alpha) = \alpha x_m^\alpha / x^{\alpha+1}, \quad x \ge x_m
$$

#### Propiedades:
- Cola pesada (no tiene media si $\alpha \le 1$).  
- No convexa ni log-cóncava.  
- Solo tratable mediante simulación o enfoque robusto.

#### Uso:
- Riesgos extremos, pérdidas financieras, tiempos de reparación largos.

---

### **8️⃣ t de Student**

#### Definición:
$$
f(x;\nu) = \frac{\Gamma((\nu+1)/2)}{\sqrt{\nu\pi}\Gamma(\nu/2)}\left(1+\frac{x^2}{\nu}\right)^{-(\nu+1)/2}
$$

#### Propiedades:
- Simétrica, colas más pesadas que la Normal.  
- Para $\nu>2$ tiene varianza finita.  
- Reformulación SOC posible (como caso elíptico).

#### Uso:
- Modelar incertidumbre con colas largas o outliers.  
- Recomendable como alternativa robusta a la Normal.

---

### **9️⃣ Cauchy**

#### Definición:
$$
f(x;x_0,\gamma) = \frac{1}{\pi\gamma[1+((x-x_0)/\gamma)^2]}
$$

#### Propiedades:
- Simétrica, colas muy pesadas.  
- No tiene media ni varianza finita.  
- Convexidad: ❌.  
- Resolución: solo simulación.

#### Uso:
- Escenarios de alta variabilidad o outliers extremos.

---

### **🔟 Gumbel (valor extremo tipo I)**

#### Definición:
$$
f(x;\mu,\beta) = \frac{1}{\beta}\exp\left(-\frac{x-\mu}{\beta} - e^{-(x-\mu)/\beta}\right)
$$

#### Propiedades:
- Modela máximos o mínimos.  
- No log-cóncava, pero cuantiles conocidos.  
- Resolución por simulación o enfoque robusto.

#### Uso:
- Picos de demanda, precipitaciones extremas, valores de máximo riesgo.

---

## 🔸 **Distribuciones discretas más comunes**

### **1️⃣ Bernoulli**

- Toma valores 0 o 1 con probabilidad $p$.  
- Convexidad trivial (conjunto discreto).  
- Se trata como escenario binario → MILP.  
- Uso: presencia/ausencia, éxito/falla.

---

### **2️⃣ Binomial**

- Número de éxitos en $n$ ensayos.  
- Soporte finito $\{0,\dots,n\}$.  
- Puede aproximarse por Normal si $n$ grande.  
- Se resuelve con escenarios o expectativas.

---

### **3️⃣ Poisson**

- Conteo de eventos por unidad de tiempo.  
- Soporte $\mathbb{N}$.  
- Cuantil numérico accesible (via `poisson.ppf`).  
- No convexa, pero simulable.

#### Uso:
- Llegadas, fallas, demandas discretas.

---

### **4️⃣ Geométrica / Negativo binomial**

- Tiempo (o cantidad de ensayos) hasta el primer éxito.  
- No convexa, pero útil para procesos con repetición.  
- Simulación o escenarios.

---

### **5️⃣ Hipergeométrica**

- Muestra sin reemplazo.  
- Convexidad no relevante (discreta finita).  
- Se maneja enumerando escenarios.

---

### **6️⃣ Discretas empíricas**

- Distribución sobre un conjunto finito $\{\xi^{(1)},...,\xi^{(N)}\}$ con pesos $p_k$.  
- Convexidad depende del peso.  
- Directamente integrable en MILP o LP si se relajan violaciones.

---

## 🔹 **Distribuciones de correlación o dependencias**

Incluso si las marginales son simples, se pueden acoplar mediante **copulas**:

| Copula | Tipo de dependencia | Aplicación |
|--------|----------------------|-------------|
| Gaussiana | simétrica, lineal | Datos normales correlacionados |
| t-Copula | colas pesadas, simétrica | Riesgos extremos conjuntos |
| Clayton | dependencias de cola inferior | Riesgos acumulativos |
| Gumbel | dependencias de cola superior | Picos simultáneos |

Estas permiten construir multivariadas realistas: por ejemplo, marginals exponenciales o lognormales acopladas por una t-Copula.

---

## 🔹 **Distribuciones truncadas**

Versiones acotadas de las anteriores (p.ej. Normal truncada en $[0, \infty)$).  
Mantienen buena convexidad, y se usan para restringir soporte físico.

---

## 🔹 **Distribuciones empíricas mixtas (de datos)**

Construidas a partir de registros o simulaciones.  
Convexidad discreta (SAA, LP o MILP), muy usadas en práctica cuando no se conoce una ley teórica.

---

## **Resumen general**

| Tipo | Ejemplos | Convexidad | Resolución típica | Comentario |
|------|-----------|-------------|------------------|-------------|
| Elípticas | Normal, t, Laplace | ✅ | SOC/LP | Base robusta |
| Log-cóncavas | Exp, Gamma, Beta(>1), Uniforme | ✅ | Convexificación / MC | Muy útil |
| Colas pesadas | Pareto, Cauchy, Lognormal | ❌ | Monte Carlo | Riesgo extremo |
| Discretas | Bernoulli, Poisson, Binomial | ⚠️ | Escenarios / MILP | Comunes en operaciones |
| Empíricas | Dataset histórico | ✅ | SAA | Basadas en datos |
| Copulas | Gaussian, t, Gumbel | ❌ | Simulación correlada | Dependencia realista |

---

### **Conclusión**

Las distribuciones más “clásicas” de teoría de probabilidades también pueden usarse en optimización estocástica.  
El criterio principal es:

- Si es **log-cóncava o elíptica**, puede tener una **reformulación convexa (LP o SOC)**.  
- Si no lo es, el camino es **simulación Monte Carlo**, **SAA**, o **enfoques robustos**.

En la práctica, vale la pena experimentar con:
- **Normal** (base teórica),  
- **t de Student** (colas pesadas),  
- **Exponencial / Gamma** (positivas),  
- **Uniforme** (acotadas),  
- **Lognormal o Pareto** (riesgo extremo),  
- **Poisson / Binomial** (demandas discretas),  
- y **Copulas** para correlaciones.

# **Hydro Planning (HP)** — Chance-Constrained con incertidumbre en el LHS (Example 2 y Sec. 6.2)

Buscamos planificar liberaciones o generación hidroeléctrica **minimizando el costo total**, garantizando con alta probabilidad que la **producción efectiva** (afectada por incertidumbre hidráulica) se mantenga dentro de una **banda operativa segura**.

---

### **Formulación (CC “pura”, usada en nuestra Etapa 1)**

**Versión con incertidumbre en el lado izquierdo (LHS):**

$$
\begin{aligned}
\min_{x \in \mathbb{R}^n} \quad & c^\top x \\
\text{s.a.} \quad & \mathbb{P}_\xi\!\big[\, A(\xi)\,x \le b \,\big] \;\ge\; 1-\varepsilon, \\
& x \in X.
\end{aligned}
$$

**Versión “two-sided” (típica en HP; dos restricciones opuestas):**

$$
\begin{aligned}
\min_{x \in \mathbb{R}^n} \quad & c^\top x \\
\text{s.a.} \quad &
\mathbb{P}_\xi\!\big[\, \ell \;\le\; A(\xi)\,x \;\le\; u \,\big] \;\ge\; 1-\varepsilon, \\
& x \in X .
\end{aligned}
$$

> *Nota:* En la Sección 6.2 del paper, el modelo HP se plantea con **incertidumbre en los coeficientes** (lado izquierdo).  
> Aquí partimos con la **Chance-Constraint estándar** (sin ambigüedad ni pesimismo), y más adelante podremos incorporar la versión robusta/pesimista.

---

### **Definición de variables y parámetros**

- **Decisión:**  
  $x \in \mathbb{R}^n$: vector de decisiones de liberación o generación (una variable por central o periodo).

- **Costo:**  
  $c \in \mathbb{R}^n_{+}$: costos unitarios asociados (energía de respaldo, bombeo, penalizaciones, etc.).

- **Productividad aleatoria (LHS):**  
  $A(\xi) \in \mathbb{R}^{m\times n}$: matriz de coeficientes de producción **aleatoria**, dependiente de condiciones hidráulicas $\xi$ (nivel, “head”, caudal, eficiencia).  
  Cada fila $a_i(\xi)^\top$ representa la energía efectiva generada bajo incertidumbre.

- **Banda/objetivo:**  
  $\ell, u \in \mathbb{R}^m$: límites inferior y superior deseados (p. ej. demanda mínima y máxima).  
  Alternativamente, si sólo hay límite superior, se usa $A(\xi)x \le b$.

- **Incertidumbre:**  
  $\xi \in \mathbb{R}^q$: vector de variables hidráulicas aleatorias (aforos, niveles, etc.).  
  **Supuesto de Etapa 1:** $\xi \sim \mathcal{N}(\mu,\Sigma)$ y $A(\xi)$ se genera como una función **afín** de $\xi$:

  $$
  A(\xi) = A_0 + \sum_{k=1}^{q} \xi_k A_k.
  $$

- **Nivel de servicio:**  
  $\varepsilon \in (0,1)$: tolerancia al riesgo; $1-\varepsilon$ es el nivel de confianza conjunto.

- **Conjunto determinista:**  
  $X$: restricciones lineales o cotas sobre $x$ (por ejemplo, $0 \le x \le U$, balances, límites de operación).

---

### **Notas de estructura (coinciden con el paper)**

- **Incertidumbre LHS:** la incertidumbre entra en los coeficientes, por lo que $\xi$ y $x$ **interactúan directamente**.  
- **Two-sided CCP:** se requiere que la producción se mantenga **dentro de una banda** ($\ell \le A(\xi)x \le u$),  
  lo que equivale a **dos** conjuntos de restricciones conjuntas (filas opuestas).  
- **Reformulaciones:**  
  - Si $\xi$ es **Normal** y $A(\xi)$ es **afín**, las restricciones **individuales** admiten una forma **SOC** exacta.  
  - La restricción conjunta completa se abordará mediante **Monte Carlo**, y compararemos las alternativas **ICC**, **SAA** y **SOC** como en PP.  
- **Interpretación:** este modelo refleja la variación estocástica de la eficiencia hidroeléctrica y de la disponibilidad de agua, buscando un plan que sea operativo con alta probabilidad.

---


---

## **Caso 1 — Configuración base para el modelo Hydro Planning (HP)**

En este primer caso reproducimos el experimento base del *Example 2 / Sec. 6.2* del paper, adaptado a nuestra Etapa 1 (sin ambigüedad ni enfoque pesimista).

### **Objetivo**

Planificar las **liberaciones de agua** o **producción hidroeléctrica** de modo que el nivel de embalse permanezca dentro de límites seguros **con alta probabilidad**, minimizando el costo total de operación.

---

### **Estructura dinámica del sistema**

- Horizonte de tiempo: $T = 5$ períodos.
- Nivel inicial: $\ell_0 = 1$.
- Evolución del nivel:
  $$
  \ell_{t+1} = \ell_t + \xi_t - x_t, \qquad t = 1,\dots,T.
  $$
  donde:
  - $\xi_t$: afluente (caudal de entrada, aleatorio),
  - $x_t$: liberación o generación (decisión de control).

---

### **Restricción probabilística conjunta**

Queremos que **todos los niveles se mantengan dentro de la banda operativa** $[\ell_{\text{low}}, \ell_{\text{high}}]$ con alta probabilidad:

$$
\mathbb{P}\Big[\, \ell_{\text{low}} \le \ell_t(x,\xi) \le \ell_{\text{high}},\; \forall\,t=1,\dots,T \Big] \;\ge\; 1 - \varepsilon.
$$

- Equivalente a un **Joint Chance Constraint (JCC)** sobre los niveles $\ell_t$.
- Se evaluará mediante **Monte Carlo**.

---

### **Parámetros del caso**

| Parámetro | Valor | Descripción |
|------------|--------|-------------|
| $T$ | 5 | Número de períodos |
| $\ell_0$ | 1 | Nivel inicial |
| $\ell_{\text{low}}$ | 1 | Límite inferior del embalse |
| $\ell_{\text{high}}$ | 5 | Límite superior del embalse |
| $\varepsilon$ | 0.10 | Nivel de riesgo (90 % de confiabilidad) |
| $U$ | 2 | Máxima liberación por período |
| $c_t$ | uniformes en $\{1,\dots,10\}$ | Costos unitarios de liberación |

---

### **Distribución de los afluentes**

Cada $\xi_t$ sigue una **Normal truncada** $\mathcal{TN}(\mu,\sigma,[0,2])$ con:

- Media $\mu = 1.0$  
- Desvío estándar $\sigma = 0.1$  
- Correlación temporal $\rho = 0.10$ entre $\xi_t$ y $\xi_{t+1}$  
- Soporte acotado $[0,2]$

> Esta elección reproduce la distribución usada en el paper: flujo medio unitario con pequeñas fluctuaciones positivas, suavemente correlacionadas entre períodos.

---

### **Resumen conceptual**

El problema combina:
- **Incertidumbre LHS:** $A(\xi)$ depende de los afluentes (coeficientes aleatorios).  
- **Estructura temporal:** la factibilidad depende del **historial acumulado** $\ell_t$.  
- **Evaluación probabilística conjunta:** el evento “banda de nivel” debe cumplirse para **todos** los períodos.

---


---

## **Método 1 — Monte Carlo Sampling (Simulación directa de la probabilidad conjunta)**

### **Motivación**

En este modelo, la restricción de probabilidad involucra el nivel del embalse en **múltiples períodos correlacionados**.  
No existe una forma analítica para la probabilidad conjunta de que todos los niveles $\ell_t$ permanezcan dentro de los límites.  

Por tanto, utilizamos el **método de Monte Carlo**:

- Generamos un gran número de trayectorias $\xi^{(k)} = (\xi^{(k)}_1,\dots,\xi^{(k)}_T)$ de los afluentes aleatorios.  
- Calculamos, para un plan candidato $x$, las trayectorias de nivel $\ell_t^{(k)}(x)$.  
- Estimamos la fracción de trayectorias que cumplen los límites en todos los períodos.  

---

### **Formulación probabilística**

Para un vector de decisiones $x=(x_1,\dots,x_T)$, definimos el evento factible:

$$
E(x,\xi) = \Big\{\, \ell_{\text{low}} \le \ell_t(x,\xi) \le \ell_{\text{high}},\; \forall t=1,\dots,T \Big\}.
$$

La restricción probabilística es:

$$
\mathbb{P}_\xi\big[\, E(x,\xi)\,\big] \;\ge\; 1-\varepsilon.
$$

La estimamos con Monte Carlo como:

$$
\widehat{p}_{\text{ok}}(x)
=
\frac{1}{N}\sum_{k=1}^{N}
\mathbf{1}\!\Big(E(x,\xi^{(k)})\Big),
\qquad
x \text{ es factible si } \widehat{p}_{\text{ok}}(x)\ge 1-\varepsilon.
$$

---

### **Por qué usar Monte Carlo aquí**

- La distribución **truncada y correlacionada** de $\xi$ no admite cuantiles cerrados.  
- La restricción conjunta acopla no linealmente los niveles entre períodos.  
- Monte Carlo permite estimar directamente la probabilidad conjunta sin derivar expresiones analíticas.  
- Es el método que el propio paper usa para estimar confiabilidad *out-of-sample*.

---

### **Ventajas**

✅ Conceptualmente simple.  
✅ Exacto en el límite de $N\to\infty$.  
✅ Requiere solo un generador de muestras de $\xi$.  
✅ Permite extender fácilmente a distribuciones no normales.

---

### **Desventajas**

⚠️ No provee forma convexa analítica.  
⚠️ Costoso computacionalmente (especialmente si $N$ grande o $T$ largo).  
⚠️ Introduce error estadístico; requiere control de la semilla y tamaño muestral.

---

### **Resumen del método**

1. Generar $N$ trayectorias correlacionadas de $\xi$ (Normal truncada).  
2. Para cada trayectoria, simular $\ell_t$ dado $x$.  
3. Contar cuántas cumplen los límites $[\ell_{\text{low}},\ell_{\text{high}}]$ en todos los períodos.  
4. Calcular la fracción de cumplimiento y verificar si es ≥ $1-\varepsilon$.  
5. Buscar $x$ que minimice el costo $c^\top x$ y satisfaga esa condición.

---


```python
# === MÉTODO 1: Monte Carlo Sampling para HP (Caso 1) ===
import numpy as np
from scipy.stats import truncnorm

# --- Parámetros del caso (como en el markdown) ---
T = 5
ell0 = 1.0
ell_low, ell_high = 1.0, 5.0
eps = 0.10
U = 2.0
c = np.arange(1, T+1).astype(float)  # ejemplo simple: costos crecientes

mu = 1.0
sigma = 0.1
rho = 0.10  # correlación entre afluentes consecutivos

# --- Construcción de matriz de covarianza correlacionada ---
Sigma = sigma**2 * (rho ** np.abs(np.subtract.outer(np.arange(T), np.arange(T))))

# --- Generador de muestras truncadas correlacionadas ---
def sample_truncnorm_correlated(mu, Sigma, N, a=0, b=2):
    """Muestrea N vectores ~ Normal(mu,Sigma) truncados a [a,b]^T"""
    L = np.linalg.cholesky(Sigma)
    z = np.random.randn(N, T)
    xi = mu + z @ L.T
    return np.clip(xi, a, b)

# --- Simulación Monte Carlo de probabilidad conjunta ---
def prob_factibilidad(x, N=10000, seed=123):
    rng = np.random.default_rng(seed)
    xi_samples = sample_truncnorm_correlated(mu, Sigma, N)
    ok = 0
    for xi in xi_samples:
        ell = ell0
        factible = True
        for t in range(T):
            ell = ell + xi[t] - x[t]
            if (ell < ell_low) or (ell > ell_high):
                factible = False
                break
        ok += int(factible)
    return ok / N

# --- Evaluación de factibilidad para un candidato x ---
x_test = np.full(T, 1.0)  # liberaciones promedio = 1
p_ok = prob_factibilidad(x_test, N=10000)
print(f"Probabilidad estimada de cumplimiento conjunto: {p_ok:.4f}")
print(f"Nivel de servicio requerido: {1 - eps:.2f}")

# --- Búsqueda simple del mejor x (baja dimensionalidad) ---
grid = np.linspace(0.5, 1.5, 21)
best_x, best_cost, best_p = None, np.inf, 0
for q in grid:
    x = np.full(T, q)
    p_ok = prob_factibilidad(x, N=5000)
    cost = c @ x
    if p_ok >= 1 - eps and cost < best_cost:
        best_x, best_cost, best_p = x, cost, p_ok
    print(f"x={q:.2f}, p_ok={p_ok:.3f}, cost={cost:.2f}")

print("\n--- RESULTADOS ---")
print(f"x* = {best_x}")
print(f"Costo óptimo = {best_cost:.2f}")
print(f"Nivel de servicio ≈ {best_p:.3f}")
```

## **Método 2 — Bonferroni / Individual Chance Constraints (ICC)**

### **Motivación**

En el modelo *Hydro Planning*, la restricción de probabilidad conjunta
$$
\mathbb{P}\!\Big[\ell_{\text{low}} \le \ell_t(x,\xi) \le \ell_{\text{high}},\; \forall t=1,\dots,T\Big] \ge 1-\varepsilon
$$
acopla todos los períodos del horizonte, por lo que es **difícil de evaluar analíticamente** y requiere simulación Monte Carlo.

Una forma clásica de obtener una **aproximación determinista y convexa** es aplicar la **desigualdad de Boole (o de Bonferroni)**, reemplazando la restricción conjunta por un conjunto de **restricciones individuales** que garanticen el mismo nivel de confianza global.

---

### **Idea principal**

En lugar de exigir que el nivel del embalse se mantenga simultáneamente dentro de la banda operativa en todos los períodos, exigimos que se mantenga con alta probabilidad en cada período individual:

$$
\mathbb{P}\!\Big[\ell_t(x,\xi)\in[\ell_{\text{low}},\ell_{\text{high}}]\Big]\ge 1-\varepsilon_t,
\quad t=1,\dots,T,
$$

sujeto a  
$$
\sum_{t=1}^{T}\varepsilon_t \le \varepsilon.
$$

Esta aproximación se denomina **Individual Chance Constraints (ICC)**.

> Si cada restricción individual se cumple con probabilidad $1-\varepsilon_t$,  
> entonces la probabilidad conjunta de cumplimiento es al menos $1-\sum_t\varepsilon_t\ge 1-\varepsilon$.

---

### **Aplicación al modelo HP**

Recordemos que los niveles evolucionan como:

$$
\ell_t = \ell_0 + \sum_{i=1}^t (\xi_i - x_i).
$$

Aproximando $\xi_i$ como **Normal** (ignorando la truncación),  
tenemos que $\ell_t \sim \mathcal N(m_t, s_t^2)$ con:
$$
m_t = \ell_0 + t\mu - \sum_{i=1}^t x_i, \qquad
s_t^2 = \mathbf{1}_t^\top \Sigma_{1:t,1:t} \mathbf{1}_t.
$$

Las restricciones individuales se traducen en condiciones sobre los **cuantiles normales**:
$$
\begin{aligned}
m_t + s_t z_{1-\varepsilon_t^+} &\le \ell_{\text{high}}, \\
m_t - s_t z_{1-\varepsilon_t^-} &\ge \ell_{\text{low}},
\end{aligned}
$$
donde $z_{1-\alpha} = \Phi^{-1}(1-\alpha)$ y dividimos el riesgo como $\varepsilon_t = \varepsilon_t^+ + \varepsilon_t^-$.

---

### **Asignación simple de riesgos**

Para simplificar:
$$
\varepsilon_t = \frac{\varepsilon}{T},
\qquad
\varepsilon_t^+ = \varepsilon_t^- = \frac{\varepsilon}{2T}.
$$

Así, los límites sobre las liberaciones acumuladas son:
$$
\begin{aligned}
\sum_{i=1}^t x_i &\ge \ell_0 + t\mu - \ell_{\text{high}} + s_t z_{1-\varepsilon_t^+}, \\
\sum_{i=1}^t x_i &\le \ell_0 + t\mu - \ell_{\text{low}}  - s_t z_{1-\varepsilon_t^-}.
\end{aligned}
$$

Estas restricciones son **lineales en $x$**, por lo que el problema se convierte en un **programa lineal (LP)**.

---

### **Problema determinista resultante (LP)**

$$
\begin{aligned}
\min_{x\in\mathbb{R}^T} \quad & c^\top x \\
\text{s.a.} \quad
& L_t \le \sum_{i=1}^t x_i \le U_t, \quad t=1,\dots,T, \\
& 0 \le x_t \le U, \quad t=1,\dots,T,
\end{aligned}
$$

donde los límites acumulados $L_t, U_t$ provienen de los cuantiles anteriores.

---

### **Ventajas**

✅ Convexo y muy eficiente computacionalmente (LP puro).  
✅ Permite una interpretación directa de los cuantiles como “márgenes de seguridad”.  
✅ Fácil de ajustar asignando más o menos riesgo por período.  
✅ Se usa frecuentemente como *benchmark determinista*.

---

### **Desventajas**

⚠️ Es **conservador**: garantiza la probabilidad conjunta usando una cota superior (Boole).  
⚠️ Puede sobredimensionar la liberación o los niveles de reserva.  
⚠️ La calidad de la aproximación depende de la correlación temporal y de la magnitud de $\varepsilon$.

---

### **Rol en nuestra Etapa 1**

- Este método nos da una **solución determinista de referencia**, rápida de obtener.  
- Validaremos su **nivel de servicio real** mediante simulación Monte Carlo (usando los afluentes truncados correlacionados del Caso 1).  
- Así podremos comparar **costo vs confiabilidad** frente al Método 1 (Monte Carlo puro) y a los siguientes métodos (SAA y SOC).

---


```python
import numpy as np
from scipy.optimize import linprog
from scipy.stats import norm

# --- Reutilizamos parámetros del Caso 1 (ajusta si cambiaste algo) ---
# T, ell0, ell_low, ell_high, eps, U, c, mu, Sigma ya deben existir.
# Si no, vuelve a definirlos aquí como en el Método 1.

def solve_icc_hp_lp(c, ell0, ell_low, ell_high, mu, Sigma, eps, U):
    """
    Bonferroni/ICC para HP (two-sided):
      sum_i^t x_i bounded por L_t y U_t derivados de cuantiles normales.
    Retorna x*, costo*, bounds acumulados, y los quantiles usados.
    """
    T = len(c)
    # Asignación de riesgo simple y simétrica
    eps_t = eps / T
    eps_minus = eps_t / 2.0
    eps_plus  = eps_t / 2.0

    # s_t = std(sum_{i=1}^t xi_i)
    s = np.zeros(T)
    for t in range(1, T+1):
        Sigma_tt = Sigma[:t, :t]
        s[t-1] = np.sqrt(np.sum(Sigma_tt))  # 1^T Sigma 1

    z_plus  = norm.ppf(1 - eps_plus)
    z_minus = norm.ppf(1 - eps_minus)

    L = np.zeros(T)
    Uacc = np.zeros(T)
    for t in range(1, T+1):
        L[t-1]   = ell0 + t*mu - ell_high + s[t-1]*z_plus
        Uacc[t-1]= ell0 + t*mu - ell_low  - s[t-1]*z_minus  # z_{1-ε^-} >0

    # Convertimos a un LP en variables x con restricciones sobre sumas acumuladas
    # Para "sum_{i=1}^t x_i >= L_t" -> -C x <= -L
    # Para "sum_{i=1}^t x_i <= U_t" ->  C x <= U
    C = np.tril(np.ones((T, T)))  # matriz acumuladora
    A_ub = np.vstack([ C, -C ])
    b_ub = np.hstack([ Uacc, -L ])

    bounds = [(0.0, U)] * T
    res = linprog(c=c, A_ub=A_ub, b_ub=b_ub, bounds=bounds, method="highs")

    if not res.success:
        raise RuntimeError("ICC-LP infeasible o solver no convergió: " + res.message)

    return res.x, res.fun, (L, Uacc), (z_minus, z_plus), s

# --- Resolver ICC y validar con Monte Carlo (truncada correlacionada real) ---
x_icc, cost_icc, (L_acc, U_acc), (z_m, z_p), s_vec = solve_icc_hp_lp(
    c=c, ell0=ell0, ell_low=ell_low, ell_high=ell_high, mu=mu, Sigma=Sigma, eps=eps, U=U
)

p_ok_icc = prob_factibilidad(x_icc, N=10000, seed=321)

print("\n--- MÉTODO 2: ICC (LP) ---")
print("x* (ICC) =", np.round(x_icc, 3))
print(f"Costo (ICC) = {cost_icc:.2f}")
print(f"Nivel de servicio empírico (MC) ≈ {p_ok_icc:.3f}   (requerido {1-eps:.2f})")
```

## **Método 3 — Scenario Approach (SAA) para HP**

### **Motivación**

La restricción probabilística conjunta
\[
\mathbb{P}\Big[\ell_{\text{low}} \le \ell_t(x,\xi) \le \ell_{\text{high}},\; \forall t=1,\dots,T\Big] \ge 1-\varepsilon
\]
no tiene forma cerrada (niveles acoplados, afluentes correlacionados y truncados).  
El **enfoque por escenarios (SAA)** reemplaza la distribución verdadera por una **muestra finita** de trayectorias
\(\{\xi^{(k)}\}_{k=1}^{N_s}\) y exige que el plan \(x\) cumpla la banda de nivel **para todas las trayectorias** del conjunto de entrenamiento.

---

### **Formulación determinista (LP)**

Para cada escenario \(k\) y cada período \(t\),
\[
\ell_t^{(k)}(x)=\ell_0 + \sum_{i=1}^t\big(\xi_i^{(k)}-x_i\big).
\]
Imponemos la **banda por escenario**:
\[
\ell_{\text{low}} \le \ell_t^{(k)}(x) \le \ell_{\text{high}}, \quad
t=1,\dots,T,\; k=1,\dots,N_s.
\]

Equivalente a **desigualdades lineales en las sumas acumuladas de \(x\)**:
\[
\boxed{
\begin{aligned}
\sum_{i=1}^t x_i &\;\ge\; \ell_0 + \sum_{i=1}^t \xi_i^{(k)} - \ell_{\text{high}},\\
\sum_{i=1}^t x_i &\;\le\; \ell_0 + \sum_{i=1}^t \xi_i^{(k)} - \ell_{\text{low}}.
\end{aligned}}
\quad\forall t,k
\]

Con cotas \(0\le x_t\le U\), esto es un **LP**.

---

### **Relación con la probabilidad \((1-\varepsilon)\)**

- **SAA “duro” (esta versión):** exige cumplir **todos** los escenarios de entrenamiento  
  \(\Rightarrow\) **riesgo empírico = 0**. Es **conservador** respecto a la meta \( \varepsilon \) (sobre todo si \(N_s\) es grande).
- Variantes (para más adelante):
  - **Permitir violaciones empíricas** \(\le \lfloor \varepsilon N_s\rfloor\) con variables binarias (MILP).
  - **Reducir escenarios** (scenario reduction) y validar fuera de muestra.
  - **CVaR/SAA**: reemplazar la *chance* por un proxy convexa (CVaR) a nivel trayectoria.

En todos los casos, la **confiabilidad real** se evalúa con **Monte Carlo out-of-sample** (igual que en el paper).

---

### **Ventajas**
✅ Totalmente **determinista**; problema **lineal** (rápido).  
✅ No requiere suponer forma paramétrica de la distribución (sirve para truncadas, mixturas, empíricas, etc.).  
✅ Muy simple de implementar.

### **Desventajas**
⚠️ **Conservador** si se exige cumplir todas las trayectorias.  
⚠️ El tamaño \(N_s\) crece el número de restricciones (\(2\,T\,N_s\)).  
⚠️ No controla directamente la probabilidad objetivo; se calibra con \(N_s\) y validación MC.

---

### **Resumen operativo**
1. Generar \(N_s\) trayectorias \(\xi^{(k)}\) (Normal truncada correlacionada).  
2. Construir el **LP** con las bandas por escenario.  
3. Resolver \( \min c^\top x \) sujeto a todas las bandas.  
4. Validar la **probabilidad conjunta** con Monte Carlo out-of-sample (mismo motor del Método 1).

---


```python
# === MÉTODO 3: Scenario Approach (SAA) para HP (versión robusta) ===
import numpy as np
from scipy.optimize import linprog

# --- 1) Generar N_s trayectorias de afluentes truncados correlacionados ---
N_s = 1500  # ajustable: más N_s -> más conservador / mayor LP
xi_scen = sample_truncnorm_correlated(mu, Sigma, N=N_s)  # shape: (N_s, T)

# --- 2) Diagnóstico de intersecciones (verifica si el SAA duro es factible) ---
bL_stack, bU_stack = [], []
for xi_k in xi_scen:
    xi_cum = np.cumsum(xi_k)
    bL_stack.append(ell0 + xi_cum - ell_high)
    bU_stack.append(ell0 + xi_cum - ell_low)
bL = np.vstack(bL_stack)
bU = np.vstack(bU_stack)
max_bL = bL.max(axis=0)
min_bU = bU.min(axis=0)
gap = min_bU - max_bL
infeas = np.any(gap < 0)

print("\n--- Diagnóstico SAA duro ---")
for t in range(T):
    print(f"t={t+1:02d}: max_bL={max_bL[t]:7.3f}, min_bU={min_bU[t]:7.3f}, gap={gap[t]:7.3f}")
if infeas:
    print("⚠️  Intersección vacía detectada en algún período -> se usará versión SAA-blando con slacks.")
else:
    print("✅ Intersecciones válidas -> se intenta resolver SAA duro estándar.")

# --- 3) Función auxiliar para construir y resolver SAA duro ---
def solve_saa_hard(xi_scen):
    T_ = T
    C = np.tril(np.ones((T_, T_)))  # matriz acumuladora
    A_ub_list, b_ub_list = [], []
    for k in range(len(xi_scen)):
        xi_k = xi_scen[k]
        xi_cum = np.cumsum(xi_k)
        bL = ell0 + xi_cum - ell_high  # lower bound
        bU = ell0 + xi_cum - ell_low   # upper bound
        A_ub_list.append( C )  # Cx <= bU
        b_ub_list.append( bU )
        A_ub_list.append(-C)   # -Cx <= -bL  (equivale a Cx >= bL)
        b_ub_list.append(-bL)
    A_ub = np.vstack(A_ub_list)
    b_ub = np.hstack(b_ub_list)
    bounds = [(0.0, U)] * T_
    res = linprog(c=c, A_ub=A_ub, b_ub=b_ub, bounds=bounds, method="highs")
    return res

# --- 4) Función auxiliar para resolver SAA-blando con slacks ---
def solve_saa_soft(xi_scen, lam=1e4):
    C = np.tril(np.ones((T, T)))
    n_x = T
    n_sl = len(xi_scen) * T
    n_su = len(xi_scen) * T
    n_vars = n_x + n_sl + n_su

    f = np.zeros(n_vars)
    f[:n_x] = c
    f[n_x:n_x+n_sl] = lam
    f[n_x+n_sl:] = lam

    A_ub_rows = []
    b_ub_vals = []

    def idx_sL(k, t): return n_x + k*T + t
    def idx_sU(k, t): return n_x + n_sl + k*T + t

    for k, xi_k in enumerate(xi_scen):
        xi_cum = np.cumsum(xi_k)
        bL = ell0 + xi_cum - ell_high
        bU = ell0 + xi_cum - ell_low
        # Restricciones inferiores (>= bL)
        for t in range(T):
            row = np.zeros(n_vars)
            row[:n_x] = -C[t]
            row[idx_sL(k,t)] = -1.0
            A_ub_rows.append(row)
            b_ub_vals.append(-bL[t])
        # Restricciones superiores (<= bU)
        for t in range(T):
            row = np.zeros(n_vars)
            row[:n_x] =  C[t]
            row[idx_sU(k,t)] = -1.0
            A_ub_rows.append(row)
            b_ub_vals.append(bU[t])

    A_ub = np.vstack(A_ub_rows)
    b_ub = np.array(b_ub_vals)
    bounds = [(0.0, U)] * n_x + [(0.0, None)] * (n_sl + n_su)
    res = linprog(c=f, A_ub=A_ub, b_ub=b_ub, bounds=bounds, method="highs")
    if not res.success:
        raise RuntimeError("SAA-soft LP no convergió: " + res.message)
    x = res.x[:n_x]
    sL = res.x[n_x:n_x+n_sl].reshape(len(xi_scen), T)
    sU = res.x[n_x+n_sl:].reshape(len(xi_scen), T)
    viol_prom = (sL + sU).mean()
    return x, c @ x, viol_prom

# --- 5) Resolver el problema (duro o blando según diagnóstico) ---
if not infeas:
    res = solve_saa_hard(xi_scen)
    if res.success:
        x_saa = res.x
        cost_saa = res.fun
        viol_prom = 0.0
        modo = "duro"
    else:
        print("⚠️ SAA duro no convergió; cambiamos a versión blanda con slacks.")
        x_saa, cost_saa, viol_prom = solve_saa_soft(xi_scen, lam=1e4)
        modo = "blando"
else:
    x_saa, cost_saa, viol_prom = solve_saa_soft(xi_scen, lam=1e4)
    modo = "blando"

# --- 6) Validación por Monte Carlo ---
p_ok_saa = prob_factibilidad(x_saa, N=20000, seed=777)

# --- 7) Resultados ---
print("\n--- MÉTODO 3: Scenario Approach (SAA) ---")
print(f"Modo de resolución: {modo.upper()}")
print("x* (SAA) =", np.round(x_saa, 3))
print(f"Costo (SAA) = {cost_saa:.2f}")
print(f"Violación promedio (slacks) = {viol_prom:.6f}")
print(f"Nivel de servicio empírico (MC) ≈ {p_ok_saa:.3f}   (requerido {1-eps:.2f})")
```

## **Método 4 — Reformulación Gaussiana tipo SOC (Second-Order Cone)**

### **Motivación**

Cuando la incertidumbre es **Normal multivariante** y entra **linealmente en el lado izquierdo**
de las restricciones, es posible convertir una *chance constraint* en una **restricción cónica de segundo orden (SOC)**.  

Este enfoque proporciona una **reformulación convexa exacta** (para restricciones individuales) y una buena aproximación para restricciones conjuntas, siendo muy útil como benchmark.

---

### **Recordatorio del modelo HP**

$$
\ell_t = \ell_0 + \sum_{i=1}^t(\xi_i - x_i), \quad
\ell_{\text{low}} \le \ell_t \le \ell_{\text{high}}.
$$

Con $\xi \sim \mathcal{N}(\mu,\Sigma)$, se tiene:

$$
\ell_t = m_t + \eta_t, \qquad
m_t = \ell_0 + t\mu - \sum_{i=1}^t x_i, \qquad
\eta_t \sim \mathcal{N}(0,s_t^2),
$$

donde $s_t^2 = \mathbf{1}_t^\top \Sigma_{1:t,1:t}\mathbf{1}_t$.

---

### **Reformulación Gaussiana**

Para cada período $t$:

$$
\begin{aligned}
\mathbb{P}(\ell_t \le \ell_{\text{high}}) \ge 1-\varepsilon_t^+
&\iff m_t + s_t z_{1-\varepsilon_t^+} \le \ell_{\text{high}}, \\[3mm]
\mathbb{P}(\ell_t \ge \ell_{\text{low}}) \ge 1-\varepsilon_t^-
&\iff -m_t + s_t z_{1-\varepsilon_t^-} \le -\ell_{\text{low}}.
\end{aligned}
$$

Cada una de estas desigualdades es una **restricción de segundo orden** (norma 2) del tipo:

$$
a(x)^\top\mu + \Phi^{-1}(1-\varepsilon)\,\|\Sigma^{1/2}a(x)\|_2 \le b.
$$

En nuestro HP (lado izquierdo lineal), el término $a(x)$ **no depende de $x$** directamente,
por lo que las SOC se **reducen a restricciones lineales** (análogas al ICC),
aunque el formalismo SOC se conserva para extensiones donde $A(\xi)$ dependa de $x$.

---

### **Problema SOC resultante (caso Normal)**

$$
\begin{aligned}
\min_{x\in\mathbb{R}^T} \quad & c^\top x \\[1mm]
\text{s.a.}\quad
& \ell_0 + t\mu - \sum_{i=1}^t x_i + s_t z_{1-\varepsilon_t^+} \le \ell_{\text{high}}, \\[2mm]
& -(\ell_0 + t\mu - \sum_{i=1}^t x_i) + s_t z_{1-\varepsilon_t^-} \le -\ell_{\text{low}}, \\[2mm]
& 0 \le x_t \le U.
\end{aligned}
$$

Con asignación simétrica:
$$
\varepsilon_t^+ = \varepsilon_t^- = \frac{\varepsilon}{2T}.
$$

---

### **Ventajas**

✅ Exacta para distribuciones Gaussianas lineales.  
✅ Convexa (SOCP o incluso LP si $A(\xi)$ no depende de $x$).  
✅ Proporciona una referencia teórica de “mejor posible sin muestreo”.  
✅ Directamente interpretable en términos de cuantiles.

---

### **Desventajas**

⚠️ Requiere supuestos de normalidad y linealidad.  
⚠️ No incluye acoplamiento temporal (la conjunción se aproxima).  
⚠️ No es exacta si la distribución real es truncada o no Normal.

---

### **Rol en nuestra Etapa 1**

Sirve como **benchmark determinista-analítico**:  
- Si la distribución es realmente Normal, su confiabilidad teórica coincide con $1-\varepsilon$.  
- Si usamos truncada o correlacionada, su desempeño real se mide con **Monte Carlo** (mismo validador que en los otros métodos).

---



```python
# === MÉTODO 4: Reformulación Gaussiana tipo SOC (Normal) para HP ===
import numpy as np
from scipy.optimize import linprog
from scipy.stats import norm

def solve_soc_hp(c, ell0, ell_low, ell_high, mu, Sigma, eps, U):
    """
    Resuelve el modelo HP con restricción gaussiana tipo SOC (analítica).
    Retorna x*, costo*, y los límites acumulados usados.
    """
    T = len(c)
    eps_t = eps / T
    eps_minus = eps_t / 2.0
    eps_plus  = eps_t / 2.0

    # Desvíos s_t
    s = np.zeros(T)
    for t in range(1, T+1):
        Sigma_tt = Sigma[:t, :t]
        s[t-1] = np.sqrt(np.sum(Sigma_tt))

    z_plus  = norm.ppf(1 - eps_plus)
    z_minus = norm.ppf(1 - eps_minus)

    # Límites acumulados
    L_acc = np.zeros(T)
    U_acc = np.zeros(T)
    for t in range(1, T+1):
        L_acc[t-1] = ell0 + t*mu - ell_high + s[t-1]*z_plus
        U_acc[t-1] = ell0 + t*mu - ell_low  - s[t-1]*z_minus

    # Formulación LP (equivalente a SOC lineal en este caso)
    C = np.tril(np.ones((T, T)))  # sumas acumuladas
    A_ub = np.vstack([ C, -C ])
    b_ub = np.hstack([ U_acc, -L_acc ])
    bounds = [(0.0, U)] * T

    res = linprog(c=c, A_ub=A_ub, b_ub=b_ub, bounds=bounds, method="highs")
    if not res.success:
        raise RuntimeError("SOC-LP no convergió: " + res.message)
    return res.x, res.fun, (L_acc, U_acc)

# --- Resolver SOC y validar con Monte Carlo (truncada real) ---
x_soc, cost_soc, (L_acc, U_acc) = solve_soc_hp(c, ell0, ell_low, ell_high, mu, Sigma, eps, U)
p_ok_soc = prob_factibilidad(x_soc, N=20000, seed=999)

print("\n--- MÉTODO 4: Reformulación Gaussiana tipo SOC ---")
print("x* (SOC) =", np.round(x_soc, 3))
print(f"Costo (SOC) = {cost_soc:.2f}")
print(f"Nivel de servicio empírico (MC) ≈ {p_ok_soc:.3f}   (requerido {1-eps:.2f})")
```

## Caso 2 — Hydro Planning (HP) con parámetros estilo paper

Instancia alineada con Example 2 / Sec. 6.2.

### Horizonte y niveles
- Períodos: $T = 5$
- Nivel inicial: $\ell_0 = 1$
- Banda operativa: $[\ell_{\mathrm{low}},\, \ell_{\mathrm{high}}] = [1,\,5]$

### Costos (precio de la energía)
Para $t = 1,\dots,T$:
$$
c_t = 10 + 5 \,\sin\!\Big(\tfrac{\pi(1-t)}{3}\Big).
$$

### Afluentes (incertidumbre)
- Distribución: **Normal truncada** en $[0,2]$ con media $\mu = 1$ y desvío $\sigma = 0.1$.
- Correlación temporal: $\mathrm{Corr}(\xi_t,\xi_{t+1}) = 0.10$ para todo $t$.
  Usaremos una covarianza tipo AR(1): $\mathrm{Corr}(\xi_i,\xi_j)=0.10^{\,|i-j|}$.

### Cotas de control
- Límite por período: $0 \le x_t \le U$ con $U=2$.

### Métrica de confiabilidad (evaluación)
Para un plan $x$, la confiabilidad es la probabilidad conjunta de que los niveles
permanezcan dentro de la banda en todos los períodos:
$$
\mathbb{P}\!\Big[\, \ell_{\mathrm{low}} \le \ell_t(x,\xi) \le \ell_{\mathrm{high}},\;\forall t=1,\dots,T \Big].
$$
Se estimará con **Monte Carlo out-of-sample** usando el mismo generador de afluentes.

---


```python
# === CASO HP-2: parámetros "como en el paper" y comparación de 4 métodos para eps in {0.10, 0.15, 0.20} ===
import numpy as np
from scipy.optimize import linprog
from scipy.stats import norm

# --------------------
# Parámetros del Caso
# --------------------
T = 5
ell0 = 1.0
ell_low, ell_high = 1.0, 5.0
U = 2.0

# Costos: c_t = 10 + 5 sin( pi(1-t)/3 ), t = 1..T
t_idx = np.arange(1, T+1)
c = 10.0 + 5.0 * np.sin(np.pi * (1 - t_idx) / 3)

# Afluentes: TN(mu=1, sigma=0.1, [0,2]), correlación adyacente 0.10 (AR(1))
mu = 1.0
sigma = 0.1
rho = 0.10  # Corr(xi_t, xi_{t+1})
# Matriz de covarianza (AR(1)): Cov[i,j] = sigma^2 * rho^|i-j|
Sigma = sigma**2 * (rho ** np.abs(np.subtract.outer(np.arange(T), np.arange(T))))

# --------------------
# Generadores y helpers
# --------------------
def sample_truncnorm_correlated(mu, Sigma, N, a=0.0, b=2.0, rng=None):
    """
    Genera N muestras ~ N(mu*1, Sigma) y las trunca a [a,b] componente a componente.
    (Aproximación por recorte; válida aquí porque la truncación es suave.)
    """
    rng = np.random.default_rng() if rng is None else rng
    L = np.linalg.cholesky(Sigma + 1e-12*np.eye(Sigma.shape[0]))
    z = rng.standard_normal((N, Sigma.shape[0]))
    xi = mu + z @ L.T
    return np.clip(xi, a, b)

def prob_factibilidad(x, N=20000, seed=1234):
    """
    Estima P( niveles dentro de [ell_low, ell_high] en todos los períodos ) por Monte Carlo.
    """
    rng = np.random.default_rng(seed)
    xi_samples = sample_truncnorm_correlated(mu, Sigma, N, rng=rng)
    ok = 0
    for k in range(N):
        xi = xi_samples[k]
        ell = ell0
        factible = True
        for t in range(T):
            ell = ell + xi[t] - x[t]
            if ell < ell_low or ell > ell_high:
                factible = False
                break
        ok += int(factible)
    return ok / N

# --------------------
# Método 1: MC + búsqueda en q (x_t = q constante)
# --------------------
def solve_mc_grid_q(eps, q_grid=None, N_eval=8000, seed0=999):
    if q_grid is None:
        q_grid = np.linspace(0.2, 1.6, 29)  # rango razonable con U=2
    best = (None, np.inf, 0.0)
    for j, q in enumerate(q_grid):
        x = np.full(T, q)
        p_ok = prob_factibilidad(x, N=N_eval, seed=seed0 + j)
        cost = c @ x
        # aceptamos si satisface el nivel, y nos quedamos con el menor costo
        if p_ok >= 1 - eps and cost < best[1]:
            best = (x, cost, p_ok)
        print(f"[MC-q] q={q:.3f} -> p_ok≈{p_ok:.3f}, cost={cost:.2f}")
    return best  # x*, cost*, p_ok

# --------------------
# Método 2: ICC (LP) — Bonferroni individual por período
# --------------------
def solve_icc_hp_lp(eps):
    eps_t = eps / T
    eps_minus = eps_t / 2.0
    eps_plus  = eps_t / 2.0

    # s_t = std( sum_{i=1}^t xi_i ) = sqrt( 1^T Sigma_tt 1 )
    s = np.zeros(T)
    for t in range(1, T+1):
        Sigma_tt = Sigma[:t, :t]
        s[t-1] = np.sqrt(np.sum(Sigma_tt))

    z_plus  = norm.ppf(1 - eps_plus)
    z_minus = norm.ppf(1 - eps_minus)

    L_acc = np.zeros(T)
    U_acc = np.zeros(T)
    for t in range(1, T+1):
        L_acc[t-1] = ell0 + t*mu - ell_high + s[t-1]*z_plus
        U_acc[t-1] = ell0 + t*mu - ell_low  - s[t-1]*z_minus

    C = np.tril(np.ones((T, T)))  # sumas acumuladas
    A_ub = np.vstack([ C, -C ])
    b_ub = np.hstack([ U_acc, -L_acc ])
    bounds = [(0.0, U)] * T

    res = linprog(c=c, A_ub=A_ub, b_ub=b_ub, bounds=bounds, method="highs")
    if not res.success:
        raise RuntimeError("ICC-LP: no convergió -> " + res.message)
    return res.x, res.fun

# --------------------
# Método 3: SAA (LP) — escenarios completos
# --------------------
def solve_saa_lp(N_s=2000, rng=None):
    rng = np.random.default_rng() if rng is None else rng
    xi_scen = sample_truncnorm_correlated(mu, Sigma, N=N_s, rng=rng)

    C = np.tril(np.ones((T, T)))
    A_ub_list, b_ub_list = [], []
    for k in range(N_s):
        xi_k = xi_scen[k]
        xi_cum = np.cumsum(xi_k)
        bL = ell0 + xi_cum - ell_high   # C x >= bL  ->  -C x <= -bL
        bU = ell0 + xi_cum - ell_low    # C x <= bU
        A_ub_list.append( C );   b_ub_list.append( bU )
        A_ub_list.append(-C );   b_ub_list.append(-bL )

    A_ub = np.vstack(A_ub_list)
    b_ub = np.hstack(b_ub_list)
    bounds = [(0.0, U)] * T

    res = linprog(c=c, A_ub=A_ub, b_ub=b_ub, bounds=bounds, method="highs")
    if not res.success:
        raise RuntimeError("SAA-LP: no convergió -> " + res.message)
    return res.x, res.fun

# --------------------
# Método 4: SOC (LP en este caso) — reformulación Gaussiana individual
# --------------------
def solve_soc_hp(eps):
    # Igual que ICC pero separado como "SOC" para mantener el método explícito
    return solve_icc_hp_lp(eps)

# --------------------
# Experimentos para eps in {0.10, 0.15, 0.20}
# --------------------
eps_list = [0.10, 0.15, 0.20]

rows = []
for eps in eps_list:
    print("\n" + "="*70)
    print(f"ε = {eps:.2f}  (nivel de servicio requerido = {1-eps:.2f})")

    # (1) MC + búsqueda en q
    x_mc, cost_mc, p_mc = solve_mc_grid_q(eps, q_grid=np.linspace(0.2, 1.6, 29), N_eval=8000)
    if x_mc is None:  # si no encontró, forzamos el mejor p_mc aunque no cumpla
        x_mc, cost_mc, p_mc = np.full(T, 1.6), c @ np.full(T, 1.6), prob_factibilidad(np.full(T,1.6), N=15000)
    print(f"[MC]   best cost={cost_mc:.2f}, p_ok≈{p_mc:.3f}, x*={np.round(x_mc,3)}")

    # (2) ICC
    x_icc, cost_icc = solve_icc_hp_lp(eps)
    p_icc = prob_factibilidad(x_icc, N=20000, seed=2024)
    print(f"[ICC]  cost={cost_icc:.2f}, p_ok≈{p_icc:.3f}, x*={np.round(x_icc,3)}")

    # (3) SAA
    x_saa, cost_saa = solve_saa_lp(N_s=1500)
    p_saa = prob_factibilidad(x_saa, N=20000, seed=3030)
    print(f"[SAA]  cost={cost_saa:.2f}, p_ok≈{p_saa:.3f}, x*={np.round(x_saa,3)}")

    # (4) SOC
    x_soc, cost_soc = solve_soc_hp(eps)
    p_soc = prob_factibilidad(x_soc, N=20000, seed=4040)
    print(f"[SOC]  cost={cost_soc:.2f}, p_ok≈{p_soc:.3f}, x*={np.round(x_soc,3)}")

    rows.append(["MC+q", eps, cost_mc, p_mc])
    rows.append(["ICC",  eps, cost_icc, p_icc])
    rows.append(["SAA",  eps, cost_saa, p_saa])
    rows.append(["SOC",  eps, cost_soc, p_soc])

# --- Resumen tabular simple ---
print("\n" + "-"*70)
print(f"{'method':<8} {'eps':>5} {'cost':>12} {'p_ok':>10} {'service':>10}")
for m, e, cost, p in rows:
    print(f"{m:<8} {e:>5.2f} {cost:>12.2f} {p:>10.4f} {1-p:>10.4f}")
```

## Caso 3 — Hydro Planning (HP) de mayor escala

Queremos analizar cómo se comportan los métodos cuando el problema tiene **más períodos** y **mayor variabilidad** en los afluentes.  
La estructura es la misma del Caso 2, pero con un horizonte extendido y una incertidumbre más intensa.

---

### Horizonte y niveles
- Períodos: $T = 15$
- Nivel inicial: $\ell_0 = 1$
- Banda operativa: $[\ell_{\mathrm{low}},\, \ell_{\mathrm{high}}] = [1,\,8]$

---

### Costos (precio de la energía)
Costos no constantes en el tiempo, oscilantes y suavemente crecientes:
$$
c_t = 10 + 3 \sin\!\Big(\tfrac{\pi(1-t)}{5}\Big) + 0.5\,t, \qquad t = 1,\dots,T.
$$

---

### Afluentes (incertidumbre)
- Distribución: **Normal truncada** $\mathcal{TN}(\mu,\sigma,[0,3])$  
- Media $\mu = 1.2$, desvío $\sigma = 0.25$  
- Correlación temporal: $\mathrm{Corr}(\xi_t,\xi_{t+1}) = 0.20$  
  (se usa una estructura AR(1): $\mathrm{Corr}(\xi_i,\xi_j)=0.20^{|i-j|}$)  
- Soporte acotado $[0,3]$ (mayor rango de caudales posibles)

---

### Cotas de control
- Límite de descarga o generación: $0 \le x_t \le U$, con $U = 3.0$.

---

### Métrica de confiabilidad
Para un plan $x$, la confiabilidad es:
$$
\mathbb{P}\!\Big[\, \ell_{\mathrm{low}} \le \ell_t(x,\xi) \le \ell_{\mathrm{high}},\;\forall t=1,\dots,T \Big].
$$
La estimaremos mediante **Monte Carlo out-of-sample**, igual que en los casos anteriores.

---

### Objetivo del experimento
1. Comparar los cuatro métodos (Monte Carlo, ICC, SAA, SOC) bajo condiciones más exigentes.  
2. Evaluar:
   - Costo total ($c^\top x$)  
   - Nivel de servicio empírico (probabilidad conjunta estimada)
3. Observar la **escalabilidad** y el **comportamiento conservador** al aumentar $T$ y la correlación.

---


```python
# === CASO 3: Hydro Planning de mayor escala ===
import numpy as np
from scipy.optimize import linprog
from scipy.stats import norm

# --------------------
# Parámetros del Caso 3
# --------------------
T = 15
ell0 = 1.0
ell_low, ell_high = 1.0, 8.0
U = 3.0

# Costos
t_idx = np.arange(1, T+1)
c = 10.0 + 3.0 * np.sin(np.pi * (1 - t_idx) / 5) + 0.5 * t_idx

# Afluentes truncados correlacionados
mu = 1.2
sigma = 0.25
rho = 0.20  # correlación entre adyacentes
Sigma = sigma**2 * (rho ** np.abs(np.subtract.outer(np.arange(T), np.arange(T))))

# --------------------
# Funciones auxiliares
# --------------------
def sample_truncnorm_correlated(mu, Sigma, N, a=0.0, b=3.0, rng=None):
    rng = np.random.default_rng() if rng is None else rng
    L = np.linalg.cholesky(Sigma + 1e-10*np.eye(Sigma.shape[0]))
    z = rng.standard_normal((N, Sigma.shape[0]))
    xi = mu + z @ L.T
    return np.clip(xi, a, b)

def prob_factibilidad(x, N=30000, seed=42):
    rng = np.random.default_rng(seed)
    xi_samples = sample_truncnorm_correlated(mu, Sigma, N, rng=rng)
    ok = 0
    for xi in xi_samples:
        ell = ell0
        factible = True
        for t in range(T):
            ell = ell + xi[t] - x[t]
            if ell < ell_low or ell > ell_high:
                factible = False
                break
        ok += int(factible)
    return ok / N

# --------------------
# Métodos 1–4 (como antes)
# --------------------
def solve_mc_grid_q(eps, q_grid=None, N_eval=6000):
    if q_grid is None:
        q_grid = np.linspace(0.3, 2.0, 35)
    best = (None, np.inf, 0.0)
    for j, q in enumerate(q_grid):
        x = np.full(T, q)
        p_ok = prob_factibilidad(x, N=N_eval, seed=100+j)
        cost = c @ x
        if p_ok >= 1 - eps and cost < best[1]:
            best = (x, cost, p_ok)
        print(f"[MC] q={q:.2f}, p_ok={p_ok:.3f}, cost={cost:.2f}")
    return best

def solve_icc_hp_lp(eps):
    eps_t = eps / T
    eps_minus = eps_t / 2.0
    eps_plus  = eps_t / 2.0
    s = np.zeros(T)
    for t in range(1, T+1):
        Sigma_tt = Sigma[:t, :t]
        s[t-1] = np.sqrt(np.sum(Sigma_tt))
    z_plus  = norm.ppf(1 - eps_plus)
    z_minus = norm.ppf(1 - eps_minus)
    L_acc = np.zeros(T)
    U_acc = np.zeros(T)
    for t in range(1, T+1):
        L_acc[t-1] = ell0 + t*mu - ell_high + s[t-1]*z_plus
        U_acc[t-1] = ell0 + t*mu - ell_low  - s[t-1]*z_minus
    C = np.tril(np.ones((T, T)))
    A_ub = np.vstack([ C, -C ])
    b_ub = np.hstack([ U_acc, -L_acc ])
    bounds = [(0.0, U)] * T
    res = linprog(c=c, A_ub=A_ub, b_ub=b_ub, bounds=bounds, method="highs")
    if not res.success:
        raise RuntimeError("ICC-LP no convergió")
    return res.x, res.fun

def solve_saa_lp(N_s=2500, rng=None):
    rng = np.random.default_rng() if rng is None else rng
    xi_scen = sample_truncnorm_correlated(mu, Sigma, N=N_s, rng=rng)
    C = np.tril(np.ones((T, T)))
    A_ub_list, b_ub_list = [], []
    for xi_k in xi_scen:
        xi_cum = np.cumsum(xi_k)
        bL = ell0 + xi_cum - ell_high
        bU = ell0 + xi_cum - ell_low
        A_ub_list.append( C ); b_ub_list.append( bU )
        A_ub_list.append(-C); b_ub_list.append(-bL )
    A_ub = np.vstack(A_ub_list)
    b_ub = np.hstack(b_ub_list)
    bounds = [(0.0, U)] * T
    res = linprog(c=c, A_ub=A_ub, b_ub=b_ub, bounds=bounds, method="highs")
    if not res.success:
        raise RuntimeError("SAA-LP no convergió")
    return res.x, res.fun

def solve_soc_hp(eps):  # igual a ICC aquí
    return solve_icc_hp_lp(eps)

# --------------------
# Experimentos para ε = 0.10, 0.15, 0.20
# --------------------
eps_list = [0.10, 0.15, 0.20]
results = []

for eps in eps_list:
    print("\n" + "="*70)
    print(f"ε = {eps:.2f}  (nivel de servicio requerido = {1-eps:.2f})")

    # 1️⃣ Monte Carlo grid en q
    x_mc, cost_mc, p_mc = solve_mc_grid_q(eps)
    if x_mc is None:
        x_mc, cost_mc, p_mc = np.full(T, 2.0), c @ np.full(T,2.0), prob_factibilidad(np.full(T,2.0))
    print(f"[MC]   cost={cost_mc:.2f}, p_ok={p_mc:.3f}")

    # 2️⃣ ICC
    x_icc, cost_icc = solve_icc_hp_lp(eps)
    p_icc = prob_factibilidad(x_icc, N=25000)
    print(f"[ICC]  cost={cost_icc:.2f}, p_ok={p_icc:.3f}")

    # 3️⃣ SAA
    x_saa, cost_saa = solve_saa_lp(N_s=2000)
    p_saa = prob_factibilidad(x_saa, N=25000)
    print(f"[SAA]  cost={cost_saa:.2f}, p_ok={p_saa:.3f}")

    # 4️⃣ SOC
    x_soc, cost_soc = solve_soc_hp(eps)
    p_soc = prob_factibilidad(x_soc, N=25000)
    print(f"[SOC]  cost={cost_soc:.2f}, p_ok={p_soc:.3f}")

    results.append(["MC+q", eps, cost_mc, p_mc])
    results.append(["ICC",  eps, cost_icc, p_icc])
    results.append(["SAA",  eps, cost_saa, p_saa])
    results.append(["SOC",  eps, cost_soc, p_soc])

# --------------------
# Resumen
# --------------------
print("\n" + "-"*70)
print(f"{'method':<8} {'eps':>5} {'cost':>12} {'p_ok':>10} {'service':>10}")
for m, e, cost, p in results:
    print(f"{m:<8} {e:>5.2f} {cost:>12.2f} {p:>10.4f} {1-p:>10.4f}")
```
