# **Descripción del modelo: Finanzas evolutivas con pagos de activos endógenos**

La idea central es que los payoffs son endógenos, esto es que dependen del estado del mundo y también de cuánta riqueza del mercado se invierte en cada activo.


## 1) Mercado en tiempo discreto y activos “short-lived”
- Tiempo discreto: $t=0,1,2,\dots$.
- Hay $K\ge 2$ activos, de **vida corta**: cada activo vive **un período**, paga al final del período y luego “renace” idéntico en el siguiente.
- Cada activo $k$ tiene **oferta total constante** $V_k$.

**Intuición:** cada fecha es una “ronda”: se forman portafolios, se liquidan payoffs y se reinicia con la misma estructura.

---

## 2) Inversionistas y riqueza
- Hay $N>2$ inversionistas.
- Riqueza inicial: $w_{i0}>0$.
- Riqueza del inversionista $i$ en $t$: $w_{it}$.

**Intuición:** el desempeño se evalúa por acumulación de riqueza y, sobre todo, por **participación relativa** de riqueza.

---

## 3) Estado del mundo $s_t$ (incertidumbre exógena)
- $s_t$ es el “estado del mundo” en $t$.
- $(s_t)$ es una secuencia **i.i.d.** en un espacio medible $S$.

**Intuición:** shocks aleatorios afectan el entorno; aquí se asume sin memoria (i.i.d.) para facilitar el análisis.

---

## 4) Estrategias de inversión: pesos constantes $\lambda_i$
Cada inversionista $i$ usa una estrategia fija
$$
\lambda_i=(\lambda_{i1},\dots,\lambda_{iK}) \in \Delta^K,
$$
donde $\lambda_{ik}\ge 0$ y $\sum_{k=1}^K \lambda_{ik}=1$.

**Interpretación:** $i$ asigna cada período la misma fracción $\lambda_{ik}$ de su riqueza al activo $k$.  
No depende de $t$ ni de $s_t$: compiten **reglas fijas**.

---

## 5) Asignación agregada y fracciones de riqueza por activo
Dos agregados relevantes:

1. **Riqueza total invertida en el activo $k$ en $t$:**
$$
w_{t,k}=\sum_{i=1}^N \lambda_{ik}\, w_{it}.
$$

2. **Riqueza total del mercado en $t-1$:**
$$
W_{t-1}=\sum_{i=1}^N w_{i,t-1}.
$$

y la **fracción de riqueza** invertida en el activo $k$ (con información de $t-1$):
$$
r_{t-1,k}=\frac{\sum_{i=1}^N \lambda_{ik}\, w_{i,t-1}}{W_{t-1}} \in [0,1].
$$
**Intuición:** $r_{t-1,k}$ mide qué tan “poblado” queda el activo $k$. Esto crea interacción: si muchos se van al mismo activo, su desempeño puede cambiar.

---

## 6) Payoffs endógenos $A_{t,k}$
El payoff del activo $k$ en $t$ es
$$
A_{t,k}=D_k\!\big(s_t,\, r_{t-1,k}\big)\, W_{t-1},
\qquad D_k(s,b)>0,\; b\in[0,1].
$$

**Puntos clave:**
- Depende del shock $s_t$.
- Depende de la **fracción de riqueza** invertida en el activo $k$ (endogeneidad/congestión).
- Es **lineal en** $W_{t-1}$. Por eso, muchas comparaciones relevantes son **relativas**.

**Consecuencia típica:** ratios como $A_{t,k}/A_{t,m}$ dependen de asignaciones relativas (el factor $W_{t-1}$ se cancela).

---

## 7) Precios por market clearing (oferta fija)
El precio del activo $k$ en $t$ queda determinado por clearing:
$$
p_{t,k}=\frac{1}{V_k}\sum_{i=1}^N \lambda_{ik}\, w_{it}.
$$

**Intuición:** si colectivamente se destina más riqueza al activo $k$, su precio sube porque la oferta $V_k$ es fija.

---

## 8) Portafolio en unidades y dinámica de riqueza
Si $x_{i,t,k}$ es la cantidad (unidades) del activo $k$ que compra $i$ en $t$, entonces:
$$
x_{i,t,k}=\frac{\lambda_{ik} w_{it}}{p_{t,k}}.
$$

La riqueza en $t$ proviene del payoff del portafolio armado en $t-1$:
$$
w_{it}=\langle A_t,\, x_{i,t-1}\rangle
      =\sum_{k=1}^K A_{t,k}\, x_{i,t-1,k}.
$$

**Intuición:** cada período es “comprar hoy con tu riqueza” y “cobrar mañana”; esa riqueza nueva determina tu poder en la siguiente ronda.

---

## 9) Admisibilidad (precios estrictamente positivos)
Un perfil de estrategias es **admisible** si, para cada activo $k$,
$$
\sum_{i=1}^N \lambda_{ik}>0.
$$

**Interpretación:** siempre hay demanda agregada positiva por cada activo, por lo que los precios quedan bien definidos y positivos.

---

## 10) Objeto evolutivo natural: participación de mercado
Con lo anterior, es natural estudiar la **riqueza relativa** (market share) del inversionista $i$:
$$
r_{it}=\frac{w_{it}}{W_t}.
$$

**Idea:** en finanzas evolutivas interesa quién **sobrevive** (no desaparece en participación) y quién domina en el largo plazo.

```python
# Colab: Escenario numérico (Parte 1 del modelo)
# Evolutive finance con payoffs endógenos

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt




# 0) Parámetros del escenario
seed = 7
rng = np.random.default_rng(seed)

T = 60          # número de períodos
N = 3           # inversionistas
K = 2           # activos

# Oferta fija de activos (V_k > 0)
V = np.array([1.0, 1.0])   # (K,)

# Estrategias fijas lambda_{ik} (cada fila suma 1)
# Asegúrate de que para cada k: sum_i lambda_{ik} > 0 (admisibilidad)
Lambda = np.array([
    [0.80, 0.20],  # Inversionista 1
    [0.50, 0.50],  # Inversionista 2
    [0.20, 0.80],  # Inversionista 3
], dtype=float)

# Riqueza inicial (normalizaremos para evitar explosión numérica)
w0 = np.array([1.0, 1.0, 1.0], dtype=float)


# 1) Estado del mundo s_t (i.i.d.)

# Opción A: Normal(0,1)
s = rng.normal(loc=0.0, scale=1.0, size=T)

# (Si prefieres estados discretos i.i.d., comenta lo de arriba y usa algo así:)
# states = np.array([-1.0, 0.0, 1.0])
# probs  = np.array([0.25, 0.50, 0.25])
# s = rng.choice(states, size=T, p=probs)



# 2) Función D_k(s,b) (payoff endógeno, siempre positivo)

# Queremos D_k(s,b) > 0 y que dependa de:
# - shock s_t
# - fracción b = r_{t-1,k} invertida en el activo k
#
# Ejemplo:
# D_k(s,b) = exp(a_k * s) / (1 + c_k * b)
#  - exp(...) siempre positivo
#  - el denominador castiga "congestión" (si b sube, D baja)

a = np.array([0.35, 0.25])   # sensibilidad al estado (K,)
c = np.array([1.20, 0.80])   # congestión (K,)

def D(s_t, b_vec):
    """
    s_t: escalar
    b_vec: (K,) con b_k en [0,1]
    retorna: (K,) con D_k(s_t, b_k) > 0
    """
    return np.exp(a * s_t) / (1.0 + c * b_vec)


# 3) Simulación (usando normalización de riqueza total)

# Nota importante:
# En el paper, A_{t,k} = D_k(s_t, r_{t-1,k}) * W_{t-1}.
# Si multiplicas por W_{t-1}, los niveles w_it pueden explotar.
# Para simular de forma estable y enfocarte en dinámica relativa,
# normalizamos cada período para que sum_i w_it = 1.
# Esto preserva las participaciones r_it = w_it / W_t.

# Chequeo de admisibilidad
if not np.all(Lambda.sum(axis=0) > 0):
    raise ValueError("Perfil NO admisible: existe un activo k con sum_i lambda_{ik} = 0.")

# Normaliza riqueza inicial
w = w0 / w0.sum()

# Para guardar resultados
cols_w = [f"w{i+1}" for i in range(N)]
cols_r = [f"r{i+1}" for i in range(N)]
cols_p = [f"p{k+1}" for k in range(K)]
cols_b = [f"b{k+1}" for k in range(K)]
cols_A = [f"A{k+1}" for k in range(K)]

hist = []

for t in range(T):
    # Fracción de riqueza invertida en cada activo (r_{t-1,k} en el texto)
    # Con riqueza total normalizada a 1: b_k = sum_i lambda_{ik} w_i
    b = Lambda.T @ w  # (K,)

    # Precios por clearing: p_{t,k} = (1/V_k) sum_i lambda_{ik} w_i = b_k / V_k
    p = b / V

    # Unidades compradas: x_{i,k} = (lambda_{ik} w_i)/p_k
    # (K,) p debe ser > 0 si admisible y w > 0
    x = (Lambda * w[:, None]) / p[None, :]  # (N,K)

    # Payoffs endógenos (en unidades normalizadas; omite W_{t-1} porque lo renormalizamos)
    A = D(s[t], b)  # (K,)

    # Actualización de riqueza: w_i,t+1 = sum_k A_k * x_{i,k}
    w_new = x @ A  # (N,)

    # Renormaliza para que sum_i w = 1 (evita explosión/underflow)
    w_new = w_new / w_new.sum()
    w = w_new

    # Guarda
    row = {"t": t+1, "s_t": s[t]}
    for i in range(N):
        row[cols_w[i]] = w[i]
        row[cols_r[i]] = w[i]  # aquí w == r, por normalización
    for k in range(K):
        row[cols_b[k]] = b[k]
        row[cols_p[k]] = p[k]
        row[cols_A[k]] = A[k]
    hist.append(row)

df = pd.DataFrame(hist)





# 4) Resultados (tablas)

print("Primeras filas:")
display(df.head(10))

print("\nÚltimas filas:")
display(df.tail(10))

# Participación final
final_shares = df[[f"r{i+1}" for i in range(N)]].iloc[-1]
print("\nParticipaciones finales (r_iT):")
print(final_shares)



# 5) Gráficos (sin seaborn)

# (a) Participaciones de riqueza (r_it)
plt.figure(figsize=(10, 4))
for i in range(N):
    plt.plot(df["t"], df[f"r{i+1}"], label=f"Inversionista {i+1}")
plt.xlabel("t")
plt.ylabel("r_it (participación de riqueza)")
plt.title("Dinámica evolutiva: participaciones de riqueza")
plt.legend()
plt.grid(True)
plt.show()

# (b) Precios
plt.figure(figsize=(10, 4))
for k in range(K):
    plt.plot(df["t"], df[f"p{k+1}"], label=f"Activo {k+1}")
plt.xlabel("t")
plt.ylabel("p_tk (precio)")
plt.title("Precios por market clearing")
plt.legend()
plt.grid(True)
plt.show()

# (c) Fracción invertida en cada activo (b_k = r_{t-1,k})
plt.figure(figsize=(10, 4))
for k in range(K):
    plt.plot(df["t"], df[f"b{k+1}"], label=f"b{k+1}")
plt.xlabel("t")
plt.ylabel("b_k (fracción de riqueza invertida)")
plt.title("Asignación agregada por activo (b_k)")
plt.legend()
plt.grid(True)
plt.show()

# (d) Payoffs endógenos A_tk
plt.figure(figsize=(10, 4))
for k in range(K):
    plt.plot(df["t"], df[f"A{k+1}"], label=f"A{k+1}")
plt.xlabel("t")
plt.ylabel("A_tk (payoff)")
plt.title("Payoffs endógenos (dependen de s_t y b_k)")
plt.legend()
plt.grid(True)
plt.show()



# 6) Sugerencias rápidas de experimentos

# - Cambia Lambda para ver "selección" (quién domina).
# - Sube c_k para castigar más la congestión y favorecer coexistencia.
# - Cambia la distribución de s_t (discreta vs normal) y compara.
```
