```python
import pyomo.environ as pyo
from pyomo.environ import SolverFactory

# Generar datos
data = create_dummy_data()
opt = SolverFactory('glpk')

# --- Adaptive ---
m1 = build_adaptive_model(data)
opt.solve(m1)
print("✅ Adaptive Model solved")
print("Objective Adaptive:", pyo.value(m1.objective))

print("\n==============================\n")

# --- Non-Adaptive ---
m2 = build_non_adaptive_model(data)
opt.solve(m2)
print("✅ Non-Adaptive Model solved")
print("Objective Non-Adaptive:", pyo.value(m2.objective))
```
